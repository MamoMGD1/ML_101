###**Data Types**

To work with data properly, you must understand **what kind of data** you're dealing with.

####**1. Numerical Data**

This is **data that consists of numbers**. You can do mathematical operations on it (like adding, averaging, comparing, etc.).

* **Discrete Data**: Data that can only take **specific values** — usually whole numbers.
  
  *Examples*:

  * Number of students in a class
  * Number of pets you own
  * How many cups of coffee you drank

* **Continuous Data**: Data that can take **any value within a range**, often measured rather than counted.
  
  *Examples*:

  * Your height (like 173.5 cm)
  * Body temperature (like 36.6°C)
  * Time taken to complete a task (like 2.34 seconds)

#### **2. Categorical Data**

This is **non-numerical data** — it represents categories or labels.

* **Ordinal Data**: Data that has a **clear order or ranking**.
  
  *Examples*:

  * Education levels (High School < Bachelor's < Master's < PhD)
  * Pain scale (None < Mild < Moderate < Severe)
  * Military positions (Officer < Captain < Colonel < General)

* **Nominal Data**: Data that has **no inherent order** — it's just a label.
  
  *Examples*:

  * Blood type (A, B, AB, O)
  * People or country names (Canada, Japan, Brazil)
  * Favorite color (Blue, Red, Green)

##**Measures of Central Tendency**

Whenever we collect data from a group of people or things (a **population**), we often want to **summarize** that data with just **one single value** that represents the whole group.

But… how can a single value represent the opinions or measurements of **hundreds or even thousands** of people?

Imagine this

> You're conducting a survey about how satisfied people are with a chocolate factory's service — and the factory operates in **multiple cities**. You wouldn't want to analyze every single person's score across the country one by one.
> Instead, you'd want **one number from each city** that somehow summarizes the whole city’s opinion.

This is exactly where **measures of central tendency** come in.

They help us:

* Reduce complex data into one easy-to-understand value
* Compare groups at a glance
* Detect patterns or problems

It makes sense that this **representative value** should be **somewhere around the center** of the entire data. That’s why we call them: Measures of **Central** Tendency.

## 🧑🏻‍🏫 **Salim’s Mysterious Classroom**

Let’s dive into a real-life story to see how this plays out:

Salim is a history teacher in a local high school. He just finished grading the final exams for his class of **90 students**.

Now, according to school policy, **any student who scores low in the exams** must be given **extra academic support**, and the filters are applied by the teachers themselves.

### **Mean**

Probably the first thing that comes to everyone's mind (as well as Salim's) is the **arithmetic mean**, also known simply as the **average**.

To calculate the mean of a group of numbers:

$$
\text{Mean} = \frac{\sum_{i=1}^{n} x_i}{n}
$$

Where:

* $x_i$ = each value in the dataset
* $n$ = the number of values

So Salim does the following:

1. He calculates the **mean score** using the formula above (since it seems the rational way of filtering).
2. He splits the class into two groups:

   * Students **below the average**
   * Students **above or equal to the average**

Since there are 90 students, he naturally expects around **45 students** to fall below the average score. But…

```python
import numpy as np
import pandas as pd
```

```python
# Generate 90 random scores
n_students = 90
np.random.seed(42)
final_scores_h = np.random.randint(98,100,10)
final_scores = np.random.randint(5,25,n_students-len(final_scores_h))
final_scores = np.append(final_scores,final_scores_h)
np.random.shuffle(final_scores)

# Print the scores
print("Students' Scores:", final_scores)
```

```python
print('Mean of The Scores:',np.mean(final_scores))
print('Students Below The Mean:',np.sum(final_scores<np.mean(final_scores)))
```

Instead of 45, Salim finds that **78 students** scored **below** the average!

That’s **more than $4/5$ of the class** falling under the cutoff.
**What went wrong?**

Salim scratches his head. Something doesn't add up…

> “How can the *average* be a fair cut-off point,” he wonders,
> “if nearly *everyone* is below it?”

**🧪 What Went Wrong?**

Salim decides to inspect the actual distribution of the scores. Here's what he finds:

* Most students scored between **5 and 25**
* A small number of students scored **extremely high**, like **98 and above**

These **very high scores** are called **outliers** — values that are far away from the rest of the data.

The **arithmetic mean** is very sensitive to extreme values.
Even just **a few very high scores** can pull the average **up**, making it **misleading**.

> That’s why so many students **below the average** were actually doing *okay*, just not as well as the outliers!

---

### **Median**

Determined to be fair, Salim searches for a better way to represent the “middle” of the class.

That’s when he discovers the **median**!

The **median** is the **middle value** when the scores are arranged **in order**.

* If the number of values is **odd**, it’s the **middle one**.
* If the number is **even**, it’s the **average of the two middle values**.

###**Why is the median better here?**

Because the **median is not affected** by outliers! It gives a much better idea of what a “typical” student in the class looks like.

| Measure    | Value | Description                                                                |
| ---------- | ----- | -------------------------------------------------------------------------- |
| **Mean**   | 23.79 | Pulled down by lots of low scores and pulled *up* by outliers — misleading |
| **Median** | 16 | Represents the true middle — not affected by the top/low scorers               |

###**Final Decision**

Salim re-evaluates the class using the **median** score of **16** as the cut-off. This time, only **about half** of the class is below the threshold — a much more reasonable and fair representation.

> Moral of the story: The **mean is not always the best measure** — especially when your data has **outliers** which is told to be **skewed**!

```python
print('Median of the scores:', np.median(final_scores))
print('Students below median:',np.sum(final_scores<np.median(final_scores)))
```

**Digging Deeper — What Caused The Difference In The Class?**

Salim was now satisfied with how the **median** helped him fairly divide the students into two groups.

But he wasn't done yet.

> “**Why** is there such a big gap between the top performers and those who are struggling?” he wondered.

Determined to help his students better, Salim opened up the **student database**, looking for **common patterns**.

He suddenly noticed a problem. This time, the data wasn't about **scores** — it was about things like:

* **Preferred study time** (morning / afternoon / night)
* **Internet access at home** (yes / no)
* **Participation in extra classes** (attended / not attended)
* **Type of commute to school** (walk / bus / car)

These values are **categorical**, meaning:

* You **can’t calculate a mean** (you can’t add “car” and “bus” together!)
* You **can’t sort them numerically**, so **median doesn’t work** either

So how can Salim analyze them?

---

###**Mode**

The **mode** is the value that appears **most frequently** in a dataset.
It doesn’t care about **numerical values or order** — just **popularity**. That makes mode the perfect tool for categorical data!

```python
import pandas as pd

# Randomly generate some data for the students
np.random.seed(42)
study_times = np.random.choice(['morning', 'afternoon', 'night'], size=n_students, p=[0.2,0.4,0.4])
internet_access = np.random.choice(['yes', 'no'], size=n_students, p=[0.7, 0.3])
extra_classes = np.random.choice(['attended', 'not attended'], size=n_students, p=[0.5, 0.5])
commute_method = np.random.choice(['walk', 'bus', 'car'], size=n_students, p=[0.5, 0.4, 0.1])

# Create DataFrame
school_df = pd.DataFrame({
    'Score': final_scores,
    'Study Time': study_times,
    'Internet Access': internet_access,
    'Extra Classes': extra_classes,
    'Commute': commute_method
})

# Calculate the median
median_score = np.median(school_df['Score'])

# Split into two groups: below and above/equal to median
below_median_df = school_df[school_df['Score'] < median_score]
above_median_df = school_df[school_df['Score'] >= median_score]

# Get the mode for each group (excluding 'Score' column)
below_modes = below_median_df.drop(columns='Score').mode().iloc[0]
above_modes = above_median_df.drop(columns='Score').mode().iloc[0]

print("Most common characteristics among students BELOW the median:")
print(below_modes)

print("\nMost common characteristics among students ABOVE the median:")
print(above_modes)
```

Salim looked at the **low-performing students**, and one category **stood out**:

* Among them, the **most common commute method** was **"walk"**
* Most **high-performing students** came to school by **bus**

This seemed strange at first… but then it made sense.

> The students who **walked long distances** to school often arrived tired, late, or even missed early morning classes.

So while the **mean and median** helped Salim separate and summarize the scores, it was the **mode** that helped him **understand the underlying cause**.

> **Moral of the story:** When your data is **about labels**, **categories**, or **names**, and you want to know **what's common**, **mode** is your best friend.

##**Measures of Spread**

So far, we've learned how to represent an entire dataset with just **one number** — the center — using measures of central tendency like **mean**, **median**, and **mode**.

But here's the problem:  
*Two datasets can have the exact same central value, yet be completely different in behavior.*

Let’s say your average monthly income is $1000.

- In one case, you get **exactly $1000** every month.

- In another, you earn \$200 one month and \$1800 the next.

Both have the same average, but one is stable, and the other? Completely unpredictable.

This is where **measures of spread** come in — they help you understand:

- **How much your data fluctuates**  
- **How consistent or scattered the values are**  
- **If there are any extreme values (outliers)** that are messing with your interpretation

## 📊 **Laila and Her Business Dilemma**

Unlike Salim, who was quite an amateur in statistics, **Laila** is a bit more professional and knows how to calculate many useful statistical measures. She's the owner of a small **online business** where she sells beautiful **traditional dresses** and **handbags** to customers across the country.

To better understand her business performance and plan her growth, Laila turns to her **daily sales data** from the past three months. She wants to analyze how consistent her sales are and whether there are any patterns she should know about.

Below is the plot of her **daily sales revenue** (in dollars) over the last 3 months:

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Randomly generate sales data
np.random.seed(42)
laila_sales = np.append(np.random.randint(0,700,80),np.random.randint(700,1400,10))
laila_sales = np.where((laila_sales<100) & (np.random.rand(len(laila_sales))<0.85),350,laila_sales)

# Plotting the distribution as Histogram
plt.figure(figsize=(10, 4))
sns.histplot(laila_sales, bins=15, edgecolor='black')
plt.title("Laila's Daily Sales Distribution (Last 3 Months)", fontsize=14)
plt.xlabel("Daily Sales (USD)", fontsize=12)
plt.ylabel("Number of Days", fontsize=12)
plt.show()
```

### **Range**

For the past three months, she's been recording her **daily sales**, which range from **\$0 to \$1200 per day**.

However, when Laila plots her sales data, she notices that the **histogram looks quite noisy**. Some days she makes **almost no sales**, while on other rare days, her sales spike **well above \$1000**.

> “This data is so irregular,” she thinks.
> “How can I possibly summarize this with just a simple average?”

She remembers **Salim’s Classroom Problem**, where relying on the **mean** led to misleading conclusions due to the presence of **extreme values** (outliers). Just like in that case, Laila suspects that:

* Some **exceptionally high sales days** (e.g., during holidays or ceremonies),
* And some **very low sales days** (e.g., when her website went down),

...are outliers that don’t represent the overall trend of her business and they may lead us to inaccurate results.

The **range** of her sales (from \$0 to \$1200) gives a sense of how the data is spread but **tells her nothing about where the data is concentrated**.

One of her old friends suggests a shortcut:

> “Just cut out everything below \$100 and above \$800. Clearly, those are outliers!”

While tempting, this approach feels too rigid to Laila. She's a perfectionist and wants a method that’s:

* **Generalizable** (can work on any future data),
* **Dynamic** (adapts based on the shape of her data),
* **Mathematical** (not just based on arbitrary guesses).

Laila wants **cutoff boundaries** that automatically adjust based on the **spread** of her data:

* If the data is **tightly packed**, the cutoffs should trim **very few points**.
* If the data is **widely scattered**, the cutoffs should trim **more points**.

And that’s how Laila begins her journey.

###**Quartiles**

Suddenly, Laila realizes something important:

> “If I stop relying on hard numbers like $800 and instead base my decisions on fractions of the data, my analysis will become much more flexible.”



For example, instead of saying "remove all sales below \$100 and above \$800," she could say:

> “Remove the lowest 2% and the highest 2% of my sales.”



This approach adjusts itself based on the overall variation in her data. It's already better than choosing arbitrary thresholds.


---

####**Median and Quartiles**

While exploring this idea, Laila stumbles upon something familiar: the median.
She remembers:

The median is the value that splits the data in half—50% of the data falls below it, 50% above it.
This is also called Q2 (the second quartile).


But she learns even more:

Q1 (the first quartile) marks the point where 25% of the data falls below.

Q3 (the third quartile) marks where 75% of the data falls below.


So the data between Q1 and Q3 contains the middle 50% of her data.

> “Hmm,” Laila thinks.
“This range between Q1 and Q3 might actually represent where my typical sales lie—it excludes the extreme lows and highs.”




---

####**A Rational Cutoff?**

Still, Laila is unsure.

Cutting off data between percentiles like 2% and 98% sounds flexible—but it's not standard, and might confuse others.

On the other hand, keeping only the middle 50% (between Q1 and Q3) means ignoring half of the data—that feels too extreme!

> “There has to be a better way,” she sighs.
“Something that uses this Q1 and Q3 idea… but gives me a smarter, less drastic cutoff.”

```python
Q1,Q2,Q3 = np.quantile(laila_sales, [0.25, 0.5, 0.75])
print('Q1, Q2, Q3 Are: ',Q1,Q2,Q3)
print('Median (Q2) of The Sales: ',np.median(laila_sales))
print('Number of data falling between Q1 and Q3: ',np.sum((laila_sales>=Q1) & (laila_sales<=Q3)))
```

### **Interquartile Range (IQR)**

After understanding quartiles, Laila learns about a very practical metric in statistics called the Interquartile Range, or IQR.

$$
IQR = Q_3 - Q_1
$$


It simply measures the spread of the middle 50% of the data, giving Laila an idea of how most of the data in her dataset is scattered.

Now the question is: How can she use IQR to trim the outliers?

Laila finds a very common and reliable method:
To detect and remove outliers, use the following rule:

$$
\text{Lower Bound} = Q_1 − 1.5 × IQR
$$

$$
\text{Upper Bound} = Q_3 + 1.5 × IQR
$$


Anything below the lower bound or above the upper bound is considered an outlier, and what’s left is usually a clean, trimmed dataset she can trust for analysis.

```python
IQR = Q3-Q1
lower_bound = Q1-1.5*IQR
upper_bound = Q3+1.5*IQR
print('Number of Data Left: ',np.sum((laila_sales>=lower_bound) & (laila_sales<=upper_bound)))
print('Outliers Detected: ',laila_sales[(laila_sales<lower_bound) | (laila_sales>upper_bound)])
```

Laila, being the perfectionist she is, isn’t satisfied just yet. She wants to *test* whether this method really adjusts to new circumstances as promised. So, she makes a copy of her current, untrimmed sales data and adds a few exceptionally high numbers between **\$500 - \$1200**—imagining her business booming next month.

She applies the **IQR trimming method again** to this updated dataset to see how the new boundaries shift and whether these newly added values get flagged as outliers or not. If the method works as expected, she’ll be convinced that this isn’t just a one-time fix, but a **flexible, data-driven strategy** that adapts to change.

> Laila expects that the new high values would pull the upper boundary high enough that our old outliers **wouldn't be categorized as extreme values anymore** and are labeled as **meaningful data**.

```python
# --- Laila adds some high values (e.g., booming next month) ---
np.random.seed(42)
extended_sales = np.copy(laila_sales)
extended_sales = np.append(extended_sales, np.random.randint(500,1000,30))  # Assume next month has great sales

# Recalculate Q1, Q3, IQR for the extended data
new_Q1 = np.quantile(extended_sales, 0.25)
new_Q3 = np.quantile(extended_sales, 0.75)
new_IQR = new_Q3 - new_Q1

# New bounds
new_lower_bound = new_Q1 - 1.5 * new_IQR
new_upper_bound = new_Q3 + 1.5 * new_IQR

# Identify new outliers
new_outliers = extended_sales[(extended_sales < new_lower_bound) | (extended_sales > new_upper_bound)]

# Print Results
print("Outliers After Adding High Values: ", new_outliers,'\n')
plt.figure(figsize=(10, 4))
sns.histplot(extended_sales, bins=15, edgecolor='black')
plt.title("Laila's Daily Sales Distribution (Last 3 Months + Next Month)", fontsize=14)
plt.xlabel("Daily Sales (USD)", fontsize=12)
plt.ylabel("Number of Days", fontsize=12)
plt.show()
```

🎯 **Laila's Business Question: Dresses or Bags?**

After cleaning out the outliers and understanding the general sales pattern, Laila decides to dig a bit deeper.

She wants to know:

> “Which product gives me more **consistent** revenue—**dresses** or **handbags**?”

She splits her sales data into two parts:
- **Dress sales**

At first glance, dresses seem more profitable. But Laila has learned her lesson—**mean alone is not enough**.

📉 Step 2: Compare the Variability (Variance)

Now Laila calculates the **variance** for each product line:

- **Dress variance** is **very high**. Some days she can't even sell any dresses but some other days its profit is \$950 or even more.
- **Bag variance** is **much smaller**. Most of the time, bag sales hover around \$300.

Even though dresses bring higher average revenue, their **instability** makes them **risky**. It’s harder to predict how much she’ll earn from them on any given day.

> Bags, on the other hand, are **more reliable**. Their revenue may be lower, but it’s consistent—something every smart business owner values.

---

### **Variance**

Variance tells Laila **how far the values are spread out from the average**.  
- **Small variance** = more stable data  
- **Large variance** = more fluctuation and risk

$$
\text{Variance} = \frac{1}{n} \sum_{i=1}^{n} (x_i - \bar{x})^2
$$

> Laila learns that when it comes to planning a sustainable business, **consistency** can sometimes be more valuable than high but unpredictable profits.

```python
# Trim the outliers
laila_sales = laila_sales[(laila_sales>=lower_bound) & (laila_sales<=upper_bound)]

# Plot dress & bag sales
bound = np.quantile(laila_sales,0.5)
dress_sales = laila_sales[(laila_sales>bound) | (laila_sales<=150)]
bag_sales = laila_sales[(laila_sales<=bound) & (laila_sales>150)]
dress_var,bag_var,dress_mean,bag_mean = dress_sales.var(),bag_sales.var(),dress_sales.mean(),bag_sales.mean()
sns.histplot(dress_sales,bins=12)
plt.xlabel('Dress Sales')
plt.xlim(0,1000)
plt.show()
sns.histplot(bag_sales,bins=5)
plt.xlabel('Bag Sales')
plt.xlim(0,600)
plt.show()

# Display means & variances
laila_df = pd.DataFrame({'Goods':['Dresses','Bags'],'Means':[np.round(dress_mean,2),np.round(bag_mean,2)],'Variances':[np.round(dress_var,2),np.round(bag_var,2)]})
display(laila_df)
```

### **Standard Deviation**

Variance is useful, but its unit is in **squared dollars**, which is not very intuitive.

That’s why Laila takes the **square root of the variance**, giving her something even easier to interpret:

> **Standard Deviation**  
> It tells you the average amount by which sales deviate from the mean—**in the same unit as the original data**.

If the standard deviation of bag sales is \$67, and for dress sales is \$211, Laila can now clearly see how much uncertainty each product carries.

```python
# Calculate standard deviations and round by 2
laila_df['Std'] = np.round(dress_sales.std(),2),np.round(bag_sales.std(),2)
display(laila_df)
```

Thanks to statistics, Laila no longer runs her business on guesswork — she’s now making data-informed decisions.

## **Data Distribution**

So far, we have covered two of the most fundamental topics in statistics: **measures of central tendency** and **measures of spread**, both of which will always stay with us throughout the course.  

We noticed how we can represent an entire dataset (no matter how big it is) with:  
- a **central tendency** → showing where the "center" of the data lies,  
- and a **measure of spread** → showing how far the data tends to deviate from that center.  

Both are powerful, but a natural question arises:  

>*Now that we know where our data is centered and how much it typically deviates, is there some way of representing the **whole data behavior** as a mathematical function?*  

Think about it. How amazing would it be to have a function that not only summarizes the data, but also lets us **ask questions about likelihood**:  

- *How likely is it that something specific happens?*  
- *How unusual is a given observation compared to the rest of the data?*  

---

### **Real-Life Motivation**
As we said earlier, our lives are full of making decisions based on what we already know (**data**) and trying to infer what might come next.  

- Based on the history of your teacher canceling classes, **what’s the chance tomorrow’s exam gets canceled** so you don’t need to study tonight?  
- In Salim’s classroom, based on past exam scores, **what’s the chance a randomly chosen student will score above 90** in the next exam?  
- In Laila’s business, looking at her last 3 months of sales, **what’s the chance tomorrow’s sales exceed \$1000**?  
- In the stock market, given the fluctuations of the past few weeks, **how likely is it that today’s price is the minimum, so I should buy now?**  

To answer these types of questions rigorously, we need **data distributions**. Data distributions are the **mathematically modeled version of our data**.  

We start with a dataset, and then model it on a graph that shows us how the data is spread out. This graph is not just for looks — it gives us direct insight into **how likely or unlikely** certain outcomes are.  

Think back to Laila’s sales. Remember the **histogram plot**?  
That histogram told her how many days her sales fell into a certain range. In other words, it visually showed the **distribution of her sales data**.  

But histograms are only the surface. Behind them lies something even more fundamental, but first we have to know the **types of distribution** themselves.  

### **Uniform Distribution**

The **uniform distribution** is the simplest form of data distribution. It literally means that every outcome is **equally likely**. In such data, you can never predict which specific outcome will happen next — because all of them share the exact same probability.  

- **Example 1: Rolling a fair die 🎲**  
  The result will always be in the range $[1, 6]$. By looking at past rolls you could see no clear pattern so each number has the same probability of appearing.  

- **Example 2: Tossing a coin 🪙**  
  The two possible outcomes are $[\text{Heads}, \text{Tails}]$. Again, both have equal likelihood.  

Formally, for a uniform distribution over a range $[a, b]$:

$$
\text{Discrete: } p(x) = \frac{1}{n}, \quad x \in \{1,2,\dots,n\}
$$

$$
\text{Continuous: } p(x) = \frac{1}{b-a}, \quad a \leq x \leq b
$$

This means the probability is evenly spread across the entire interval no matter what your $x$ is.  

So if you draw a random number from a uniform distribution in the range $[1, 10]$, the probability of seeing **1**, **2**, **3**, or even **10** is always the same. Thus, after let's say 10,000 tests we could see that all the outputs have been witnessed closed to 1000 times.

```python
np.random.seed(42)
upper_bound = 10
lower_bound = 1

# Choose 10000 random numbers between lower_bound and upper_bound
outputs = np.random.uniform(lower_bound, upper_bound, 10000)

sns.histplot(outputs, bins=10)
plt.title('Uniform Distribution')
plt.xticks(np.arange(1,11))
plt.show()
```

### **Normal Distribution**

No matter how eager you are to believe that almost every random thing happening around us comes from a uniform distribution — since it often seems like all outcomes are equally likely — it's actually **not true**.  

In fact, there is another type of distribution that **a huge number of real-world phenomena follow**, and it comes with many important assumptions and properties. This distribution is called **Normal Distribution** and due to the extensive contributions of **Carl Friedrich Gauss (1777–1855)**, this distribution is also called the **Gaussian Distribution**.  

To see why we focus so much on it, let’s consider a random experiment:  

Suppose you stand on the street with a thermometer and randomly measure the **body temperature** of everyone you see — no matter their age, gender, or health status. You might expect that the temperatures will randomly fall anywhere between 35°C and 40°C (because anyone out of this range would not be walking on the street) with roughly equal probabilities. Right? **Wrong!**  

In reality, a clear pattern emerges:  
- Most people’s temperatures cluster around **37°C**.  
- Out of 10 people, about 5 might measure between **36.5°C and 37.5°C**.  
- Only a few people will fall outside this range, and you might not even see anyone at exactly 35°C or 40°C in 100 measurements.  

Clearly, the data is **not uniformly distributed**. Instead, it follows a pattern:  
- Values near the **central point** (mean) are more likely.  
- Likelihood **gradually decreases** as you move further away from the mean.  

This type of distribution — where data is spread around a central value in this bell-shaped pattern — is called the **Normal Distribution**, **Gaussian Distribution**, or simply the **Bell Curve Distribution**.

<div align="center">
  <img src="https://raw.githubusercontent.com/MamoMGD1/ML_101/main/images/S00_C00/tractor.png" width="600">
</div>

This idea of data being **randomly distributed around a central value** is something we see all the time. For example, consider the **weights of high school students**. If you enter a classroom in which everyone is roughly the same age and start measuring weights, you’ll notice a **mean value** around which most students cluster.  

- Suppose the average weight is **60 kg**.  
- It’s very likely to find students weighing between **55–65 kg**,  
- while finding someone weighing **40 kg** or **80 kg** is possible, but **much less likely**.  

This pattern arises because when we have **many independent random variables** (like individual factors contributing to a person’s weight), their combined effect naturally tends to form a **normal distribution** rather than a uniform one.  
- Even though uniform distributions also involve independent random variables, in the uniform case **every outcome must be exactly equally likely**, which rarely happens naturally.  
- In contrast, in most real-world phenomena, values naturally cluster around a central tendency due to **averaging effects**, giving rise to the bell curve.  

Another example is a **pile of sand** dropped from a tractor. The sand particles land randomly within a circular area on the ground. But soon, you’ll notice **more sand gathers near the center** and fewer at the edges — a visual demonstration of a normal distribution forming naturally.  


### **Standard Normal Distribution**

Thanks to the **normal distribution**, we can track and predict almost everything around us. For example:  
- Weather forecasts (temperature, rainfall)  
- Exam scores
- Heights and weights of populations
- Stock market returns   
- Manufacturing tolerances and measurement errors   

All of this works because we know outcomes are **not equally likely**. If everything were uniformly distributed, we could never make reliable predictions.  

---

One of the most useful techniques in working with normal distributions is **converting data to standard form**.  

The **standard normal distribution** is a normal distribution where:  
- Mean = 0  
- Standard deviation = 1  

This conversion is done using **Z-score scaling**:  

$$
z = \frac{x - \mu}{\sigma}
$$

- $x$ = original data point  
- $\mu$ = mean of the original data  
- $\sigma$ = standard deviation of the original data  

This transformation **centers the data** around 0 and **rescales the spread** to 1.  

Why is this so useful?  
The **area under the standard normal curve is 1**, which means the total probability is 1. If you want the probability that a value falls in a certain range, you just compute the **area under the curve** using a simple integration.

---

Another handy rule is the **Empirical Rule (68-95-99.7 Rule)**:  

- ~68% of values fall **within 1 standard deviation** of the mean  
- ~95% of values fall **within 2 standard deviations**  
- ~99.7% of values fall **within 3 standard deviations**  

<div align="center">
  <img src="https://raw.githubusercontent.com/MamoMGD1/ML_101/main/images/S00_C00/norm.jpg" width="500">
</div>

The math behind the normal distribution is deep and fascinating, but for the purposes of this course, this understanding is **enough to start working with it**. Soon, we’ll apply these concepts in real datasets and machine learning tasks.

```python
np.random.seed(42)

# Simulate average weights for 100 people, 10,000 times
num_people = 100
num_experiments = 10000

group_average_weights = []

for _ in range(num_experiments):
    group_weights = np.random.random(num_people) * 40 + 40  # Weights between 40-80 kg
    group_average_weights.append(group_weights.mean())

# Plot distribution of group averages
sns.histplot(group_average_weights, kde=True, color='skyblue')
plt.title('Distribution of Average Weight of Groups of 100 People')
plt.xlabel('Average Weight (kg)')
plt.ylabel('Frequency')
plt.show()

# Standardize the data (Z-score)
group_average_weights = np.array(group_average_weights)
mean_weight = group_average_weights.mean()
std_weight = group_average_weights.std()
standardized_weights = (group_average_weights - mean_weight) / std_weight

# Plot standardized weights
sns.histplot(standardized_weights, kde=True, color='salmon')
plt.title('Standardized Weights (Z-score)')
plt.xlabel('Z-score')
plt.ylabel('Frequency')
plt.show()

# Probability of falling within 1, 2, 3 stds in original weights
within_1std = np.mean((group_average_weights >= mean_weight - std_weight) &
                      (group_average_weights <= mean_weight + std_weight))
within_2std = np.mean((group_average_weights >= mean_weight - 2*std_weight) &
                      (group_average_weights <= mean_weight + 2*std_weight))
within_3std = np.mean((group_average_weights >= mean_weight - 3*std_weight) &
                      (group_average_weights <= mean_weight + 3*std_weight))

print(f"Probability that the next group's average weight is within 1 std ({mean_weight-std_weight:.2f}-{mean_weight+std_weight:.2f} kg): {within_1std:.3f}")
print(f"Probability that the next group's average weight is within 2 stds ({mean_weight-2*std_weight:.2f}-{mean_weight+2*std_weight:.2f} kg): {within_2std:.3f}")
print(f"Probability that the next group's average weight is within 3 stds ({mean_weight-3*std_weight:.2f}-{mean_weight+3*std_weight:.2f} kg): {within_3std:.3f}")
```

## **Correlation**

Have you ever wondered how tightly two things in life are connected? Can you take your time and find out how these things are related to each other?

* The number of ice creams sold and the temperature of the day
* The amount of time spent studying and exam scores
* The price of movie tickets and the number of people going to the cinema
* The speed of a car and the time it takes to reach a destination
* The number of cups of coffee you drink and your t-shirt’s color
* The weight of a backpack and the distance someone can walk comfortably
* The number of plants in a room and the number of pets the owner of that room keeps

I’m sure you’ve already noticed—some of these are directly related, others are inversely related, and a few don’t seem to have any relationship at all, well in statistics we actually have a measure for that as well. We call it **Correlation**.

Correlation is one of the most powerful tools in data analysis. It helps us measure the *strength* and *direction* of a relationship between two variables. It's more than just a pattern—it's a mathematical fingerprint of association.

> While correlation is based on **covariance** (which measures how two variables vary together), correlation goes a step further: it standardizes the values, allowing us to compare relationships across different scales.

---

###**Correlation Formula**

The most common measure is the **Pearson correlation coefficient (r)**:

$$
r = \frac{\sum (x_i - \bar{x})(y_i - \bar{y})}{\sqrt{\sum (x_i - \bar{x})^2 \sum (y_i - \bar{y})^2}}
$$

Where:

* $x_i, y_i$ are individual sample points,
* $\bar{x}, \bar{y}$ are the sample means.

The result is always between -1 and 1:

* **+1**: Strong positive relationship
* **0**: Not related at all
* **-1**: Strong negative relationship

Understanding correlation opens the door to deeper data stories—and helps you avoid jumping to misleading conclusions and most importantly tells you which data to focus more on during your researches.

Last year, Najla happily planted tomatoes in the small garden behind her house. Without giving it much thought, she used equal parts of all five different fertilizers she had — thinking, "Well, fertilizers help plants grow, so the more the merrier, right?"

It turned out that while some minerals in her fertilizers were essential for healthy tomato growth, others were harmful, weakening the plants and stunting their development.

So this year, Najla decided to get serious about regularizing her fertilizing strategy. So...

###**Najla Struggling Between Simple Methods:**

Najla, who is now super careful about her fertilizers, first reads the ratio of the minerals on the back of all the bags and notices that she is basically dealing with five bags of fertilizers enriched by, in order: Potassium ($K$), Calcium ($Ca$), Nitrogen ($N$), Ammonia ($NH_3$), and Iron($Fe$). So the only thing she needs is to check which of the substances above has the best effect on tomato growth by searching online. Seems like such an easy thing to do.

Except that... it is not.

All she finds are websites, each claiming different things — one saying Potassium is the king of ripeness, another saying it can actually hurt the plant. There’s no agreement, and she can't decide which one has been the deadliest culprit to cut its dose, and which one to increase.

But then, she finds **something that might help**:

A dataset of more than **1300 records of tomato weights**, with different amounts of all five of these minerals used as their fertilizers.

A single heavy tomato might have grown big because of a ton of other factors — maybe better sunlight, different watering, or even random chance.
```python
np.random.seed(42)

# Number of samples
n = 1320

# Generate independent variables (fertilizer components)
K = np.random.uniform(1, 10, n)        # Potassium
Ca = np.random.uniform(1, 10, n)       # Calcium (low effect)
N = np.random.uniform(1, 10, n)        # Nitrogen
NH3 = np.random.uniform(1, 10, n)      # Ammonia (negative effect)
Fe = np.random.uniform(1, 10, n)       # Iron (low effect)

# Generate dependent variable (tomato weight)
weight = (
    30 +
    6 * K +
    5 * N -
    4 * NH3 +
    np.random.normal(0, 1, n)  # small noise
) * 2

# Create DataFrame
df = pd.DataFrame({
    'K': K,
    'Ca': Ca,
    'N': N,
    'NH3': NH3,
    'Fe': Fe,
    'W_Tomato': weight
})

# Preview first 5 records of the dataset
df.head()
```

##**Healthy Training Dataset**

Before you ever train a machine learning model — before you even choose your algorithm — there's a critical decision that shapes everything: **how you collect your data**.

Whether you're analyzing a pre-built dataset or running a survey to solve a real-world problem, your first mission is clear:  
**Make sure your sample represents your population — fairly, fully, and without distortion.**

Why does this matter?

Because your model is only as good as the data it's trained on. If your sample ignores part of the population or emphasizes certain patterns too much, you're not teaching your model to "learn the truth" — you're teaching it to echo the flaws in your sampling. And this leads to errors, missed insights, and even dangerous biases.

This part is about **healthy dataset**, **bias in data collection**, and the twin threats of **overfitting and underfitting**. Together, they define what it means to work with *healthy*, *trustworthy*, and *useful* data.

###**A Perfect Training Gone Wrong**

Kabir, a bright young boy with dreams bigger than the circus tent itself, has been preparing his elephant, **Bholu**, for a spectacular performance at the **Great Grand Circus Show**. Bholu is not just any elephant — he's clever, curious, and incredibly responsive to Kabir’s teachings.

For weeks, Kabir designs a thoughtful and precise training routine. He breaks the performance down into **four elegant tricks**, teaching each with love, repetition, and clarity:

1. **Sit down** when a **chair** is shown.
2. **Eat** the **apple** when it’s offered.
3. **Start dancing** when the **trumpet** sounds.
4. **Walk in a circle** when the **spotlight** turns on.

Day after day, Bholu learns and performs flawlessly in Kabir's backyard. He never forgets a step. The trumpet blows? He starts dancing. The apple appears? Chomp. The spotlight shines? Perfect elephant pirouette.

Kabir beams with pride. His friends and family cheer.

### 🎪 **Rehearsals Begin!**

But then, **rehearsals** start — and something *unthinkable* happens. The same tricks. The same audience cues. But Bholu weirdly acts as if he’s never done the tricks before.

* A trainer holds up a **green apple** (instead of the red one Kabir always used)…
  
  Bholu **sniffs it cautiously**… then **throws it into the audience**.

* The **chair** is slightly taller than the one from training…
  
  Bholu **stares at it blankly**, turns around, and walks away.

* The **trumpet** sound is just a bit louder…
  
  Bholu flinches and **hides behind the curtain**.

  Bholu **freezes**, then **starts hopping** instead of walking in a circle.

Kabir is **devastated**.
How could this happen? He trained Bholu perfectly!
He’s smart, he learned so well, he *knew* what to do...

But now? In front of a real crowd? He looks like the **most confused, unreliable elephant ever**.

> “What went wrong?” Kabir whispers.
> “Why does Bholu act like he’s never learned a thing?”

Well, **the problem isn’t Bholu**.

In fact, what happened in the circus **doesn’t show a failure in Bholu’s learning at all**. It shows something far deeper and more important… **Bholu doesn’t understand the human world.**
He doesn’t know why a chair is for sitting, or what an apple is for — red, green, or golden. He doesn't reason like us, humans. **Just like machines**, he can memorize sequences, perform patterns, even solve math-like instructions… but he **doesn’t "know" what objects mean** unless **we show him clearly and generally enough**.

> A green apple to him could be a delicious snack...
> Or it could be a venomous jungle creature in disguise.
>
> To Bholu, **both are just "some new object."**
> Same goes for machines.

Humans don’t need hundreds of examples to understand that chairs are for sitting — we already carry **strong prior assumptions** about the world. But machines (and elephants) don’t.
They rely *entirely* on the data we feed them.

That’s why it’s not just about **feeding Bholu more data** — it’s about giving him **the right kind of data**, in the **right amount**, covering the **right patterns**, and **avoiding patterns that confuse him**.

---

If we want a machine (or Bholu) to learn:

* Not too little data (or they’ll forget everything on stage)...
* Not too much of the same kind (or they’ll panic when things change)...
* And **not biased or repetitive examples** that hide the real diversity of the world.

**Kabir is about to discover** that his training method was built with tiny cracks:
maybe he used too few types of apples…
maybe all the trumpets sounded the same…
maybe the chairs were always red and short…

Let’s follow Kabir as he begins his journey to **debug the data** and **teach Bholu the right way**.

###**Problem #1: Undiversified Training Set**

Kabir realized: he had trained Bholu on **objects that were too similar**.

* All apples were red.
* All chairs were short and wooden.
* All trumpets made the same pitch.

So what did Bholu learn?
**That red = eat. That square = sit. That one sound = dance.**

This is called **overfitting** — when a learner becomes **too attached to surface-level patterns** that don’t really matter.
He didn’t learn the *concepts*, he memorized the *decorations*.

So Kabir started changing things up.

He added:

* Green, yellow, red apples.
* Plastic chairs, beanbags, stools.
* Loud, soft, sharp, flat trumpet sounds.

He hoped this would teach Bholu what *really matters* — like the fact that **an apple is edible not because it's red, but because of its shape, smell, and texture**.

But...

Just when things seemed better, Bholu saw a **round, bright, bouncy tennis ball** and confidently swallowed it whole because according to his assumptions:

> Green Apple (round and green) = tastes good
>
> Tennis Ball (round and green) = so... it should also taste good

Kabir facepalmed.

Turns out, Bholu had **overfit again**, this time to a **single shape-based rule**, so **More variation was needed.**
More types of round things — some edible, some not.
**More examples to break the misleading pattern**.

###**Problem #2: Insufficient Training Set**

But another issue appeared…

Sometimes, Bholu just stood still. He stared at an object and **didn’t know what to do**.

* Is the yellow thing food or something else?
* Is this green apple really an apple or one of those disgusting balls again?

He couldn’t decide.
He tried strategies, failed, changed strategies again, and failed even harder.

Kabir finally saw what was going on:
Bholu hadn’t received **enough feedback**. His experience was **too shallow** to form a consistent rule.

This is called **underfitting** — when the learner **doesn’t get enough training data** to learn *any* useful pattern.

One time he was punished for eating something round.
Next time he was rewarded. And sometimes ignored.

> “How am I supposed to know what’s right?” — probably Bholu

---

### 📊 **Training That Actually Works**

So here’s what Kabir learned:

* To **avoid overfitting**, the training examples must be **varied enough** to teach real concepts — not cosmetic coincidences.
* To **avoid underfitting**, the training must be **rich and long enough** to let meaningful rules emerge.
* To teach someone who **doesn’t have your background knowledge**, you must **control the patterns they see**, otherwise they’ll make up their own weird rules (like eating tennis balls).

Kabir is now ready to rebuild the training set — one that is *balanced*, *diverse*, and *instructive*.
And Bholu? He’s ready to start learning like a machine.


###**Problem #3: Training Epochs**

At this point, Kabir had done everything he could with the data.
He had diversified it, balanced it, introduced enough variation to fight overfitting, and ensured it wasn’t so chaotic that it led to underfitting.

And yet… Bholu still behaved unpredictably.

That’s when Kabir realized:
**It’s not just about *what* you train on — it’s about *how many times* you train.**

* When he **trained too little**, Bholu barely understood the tasks.
  He made random guesses, had no strategy — classic **underfitting**.

* When he **trained too much**, Bholu began overanalyzing.
  He invented complex, irrelevant rules — like sniffing chairs or analyzing how many times the apple rotated before being offered.
  This was a clear case of **overfitting**.

In ML, this trade-off between how **biased** a model is from the correct answers and how **variantly** it is behaving is known as the **bias–variance tradeoff**:

> **High bias** → Model is too rigid → Underfitting
>
> **High variance** → Model is too sensitive → Overfitting

A good model (or Bholu) sits at a **point of equilibrium** — where it generalizes well to new inputs but also learns enough to capture the relevant signal.

This is why selecting an appropriate number of **training epochs** is critical. Too few, and the model never converges. Too many, and the model memorizes instead of generalizing.

---

### ⚖️ **The Bigger Picture: It's About Generalization, Not Memorization**

By this point, Kabir wasn’t just training Bholu —
He was performing an **iterative model tuning** process:
adjusting **data diversity**, **environmental robustness**, and **training duration** to optimize **generalization**.

And that’s the ultimate goal in machine learning.

> **The point is not to learn everything — but to learn what matters.**
> That’s what separates a functional ML model from a brittle, unreliable one.

* Designing learning environments that reflect real-world variability
