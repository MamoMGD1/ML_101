# **VECTORS & MATRICES IN ML**

Before we start coding ML models, let’s ask a simple question:

> **How does a machine *see* the world — and how can we make it *understand* ideas like we do?**

Imagine we give a machine (computer) two numbers: Object **A** weighs **2 kg** and object **B** weighs **10 kg**.  

The machine has no trouble telling which one is heavier — it’s just comparing two numbers.  
But real life is rarely that simple.  

---

## **When One Number Isn’t Enough**

Most real-world problems need **more than one number** to describe each thing.  

- If we want to compare **phones**, we can’t just use their price — we also care about **battery life**, **camera quality**, **RAM**, **storage**, and more.  

- If we want to compare **people’s health**, we can’t just use their BMI — we might look at:  
  - Height  
  - Weight  
  - Blood pressure  
  - Body temperature  
  - Average sleep hours  
  - Heart rate  
  - Cholesterol level  
  - …and so on  

Clearly, one number per entity (sample) is not enough.  
We need a way to represent **all these features together** so the machine can make a meaningful comparison.  

---

## **Say Hello to Vectors**

A **vector** is exactly that — a way to group multiple numbers into a single object.  
Each **dimension** of the vector represents one feature.  

For example, suppose we want to compare two people’s health using **three features**:  
weight (kg), height (cm), and average sleep hours.

- Person A → **$[72, 180, 6]$**  
- Person B → **$[65, 165, 8]$**  

Each person is now a **point in 3D space** — and we can measure *distances* between these points.  
If we had even more features (blood pressure, cholesterol, etc.), the vector would simply grow in length — it becomes a point in a higher-dimensional space.  

This is the first big idea of Machine Learning:  
> **Every sample in your dataset can be represented as a vector.**  

And once we represent our data this way, we can use **math** to compare, find differences, find similarities, detect anomalies, transform, and... But let’s be honest: as amazing as this sounds, it’s still not enough to give machines some of our **most powerful abilities** — like **vision**. One of the most astonishing things humans can do is **see**. We can look at an image and instantly recognize faces, objects, and emotions.  

How do we give machines this ability?  
At first, it seems impossible — an image doesn’t have “features” like height, weight, or blood pressure we can just list in a vector.  

But here’s the trick:  

An image *does* have numbers — just a lot of them. Each **pixel** in an image has an intensity value.  

Take a **grayscale image** with a width of **1080 px** and a height of **720 px**:  

- We can represent it as a table with **1080 columns** and **720 rows**.  
- Each **cell** in the table is a number between **0** and **255**:  
  - `0` → completely black  
  - `255` → completely white  
  - any value between → a shade of gray  

This table of numbers is what we call a **matrix**.

## **Say Hello to Matrices**

A **matrix** is simply a rectangular grid of numbers — but in ML, it becomes so much more:  

- It can represent **images**, as we just saw.  
- It can represent **datasets**, where each row is a vector (a sample) and each column is a feature.  
- It can represent **model parameters** and the transformations we apply during training.  

In other words:  
> **Matrices let us scale from single samples to entire worlds of data.**  

And this is just the beginning.
In this chapter, we explored how **vectors** and **matrices** give us a way to represent data, measure relationships (like **similarities** and **distances**), and perform operations like **transformations** and **weighted scoring**.
These tools are the foundation for everything that comes later in ML — from finding patterns to making smart decisions.

By the end, you’ll not just know what a vector or matrix is, you’ll *feel* why they are the building blocks of machine learning.

## **Vectors**

A **vector** is an **ordered list of numbers**, often representing a **point in space**, a **direction**, or a **set of features**. You can think of it as:

- A **column of values**:  
$$
  \mathbf{x} = \begin{bmatrix} x_1 \\ x_2 \\ x_3 \end{bmatrix}
$$

- Or as a **row**, depending on the context:  
$$
  \mathbf{x} = \begin{bmatrix} x_1, x_2, x_3 \end{bmatrix}
$$

In machine learning:
- A **single data sample** (with features like height, weight, and age) → a **vector**
- A model’s **parameters/weights** → a **vector**
- The **output of a neuron** in a neural network → a **vector**

---

### **Dimensions and Notation**

A vector with **n components** lives in **n-dimensional space** $(ℝ^n)$.  
Examples:
- $[5, 2]$ is a vector in 2D $(ℝ^2)$
- $[1.2, -0.5, 3.4]$ is a vector in 3D $(ℝ^3)$

Understanding dimensions is essential:
- **Shape mismatch** is one of the most common bugs in ML code.
- You **can’t add or multiply vectors** with incompatible dimensions.

### **Common Vector Operations**

Here are the most important operations you’ll use all the time in ML:

#### 1. **Norm (Length)**

The norm or magnitude is like the **length of the vector**:  
$$
\|\mathbf{x}\| = \sqrt{x_1^2 + x_2^2 + \dots + x_n^2}
$$
$$
\mathbf{x} = \begin{bmatrix} -1 \\ 3 \\ 5 \end{bmatrix}, \|\mathbf{x}\| = \sqrt{(-1)^2+3^2+5^2} = \sqrt{35} \approx 5.92
$$

---

#### 2. **Addition / Subtraction**

Element-wise operation:  
$$
\begin{bmatrix} 1 \\ 2 \\ 3 \end{bmatrix} + \begin{bmatrix} 4 \\ 5 \\ 6 \end{bmatrix} = \begin{bmatrix} 5 \\ 7 \\ 9 \end{bmatrix}
$$

---

#### 3. **Scalar Multiplication**

Multiply each element by a constant:  
$$
3 \cdot \begin{bmatrix} 1 \\ 2 \\ 3 \end{bmatrix} = \begin{bmatrix} 3 \\ 6 \\ 9 \end{bmatrix}
$$

---

#### 4. **Dot Product (Inner Product)**

Multiplies corresponding elements and sums the result:  
$$
\mathbf{a} \cdot \mathbf{b} = a_1b_1 + a_2b_2 + \dots + a_nb_n
$$

$$
\begin{bmatrix} 1 \\ 0 \\ 3 \end{bmatrix}
\cdot
\begin{bmatrix} -4 \\ 2 \\ 6 \end{bmatrix}
=
(1 \times -4) + (0 \times 2) + (3 \times 6)
= -4 + 0 + 18 = 14
$$

It tells us **how aligned** two vectors are:  
- If the dot product is **0**, the vectors are **orthogonal** (perpendicular).  
- Larger dot product → more similar direction.

---

#### 5. **Cosine Similarity**

Cosine similarity uses the dot product to measure **how similar two vectors are**, but it **removes the effect of their lengths** by dividing by the product of their magnitudes:  

$$
\text{cosine_similarity}(\mathbf{a}, \mathbf{b}) = \frac{\mathbf{a} \cdot \mathbf{b}}{\|\mathbf{a}\| \, \|\mathbf{b}\|}
$$

- Result ranges from **–1** to **1**.
  - **1** → vectors point in exactly the same direction (maximum similarity).
  - **0** → vectors are perpendicular (no similarity).
  - **–1** → vectors point in exactly opposite directions.

**Example 1:**  

$$
\mathbf{a} = \begin{bmatrix} 1 \\ 2 \end{bmatrix} \quad
\mathbf{b} = \begin{bmatrix} 2 \\ 4 \end{bmatrix}
$$  

Vector **b** is pointing in the **same direction** as vector **a**, it’s just twice as long.  
Since cosine similarity ignores length, their cosine similarity will be **1**.

$$
\text{cosine_similarity}(\mathbf{a}, \mathbf{b}) =
\frac{(1 \times 2) + (2 \times 4)}{\sqrt{1^2 + 2^2} \cdot \sqrt{2^2 + 4^2}}
= \frac{2 + 8}{\sqrt{5} \cdot \sqrt{20}}
= \frac{10}{\sqrt{100}} = 1
$$  

**Example 2:**  

$$
\mathbf{a} = \begin{bmatrix} 1 \\ 0 \end{bmatrix} \quad
\mathbf{b} = \begin{bmatrix} 0 \\ 1 \end{bmatrix}
$$  

Vector **a** points along the x-axis, and vector **b** points along the y-axis.  
They are **perpendicular**, so their cosine similarity must be **0**.

$$
\text{cosine_similarity}(\mathbf{a}, \mathbf{b}) =
\frac{(1 \times 0) + (0 \times 1)}{\sqrt{1^2 + 0^2} \cdot \sqrt{0^2 + 1^2}}
= \frac{0 + 0}{1 \cdot 1}
= 0
$$  

**Example 3:**  

$$
\mathbf{a} = \begin{bmatrix} 1 \\ 2 \end{bmatrix} \quad
\mathbf{b} = \begin{bmatrix} -1 \\ -2 \end{bmatrix}
$$  

Vector **b** points in the **exact opposite direction** of vector **a**, so their cosine similarity will be **–1**.

$$
\text{cosine_similarity}(\mathbf{a}, \mathbf{b}) =
\frac{(1 \times -1) + (2 \times -2)}{\sqrt{1^2 + 2^2} \cdot \sqrt{(-1)^2 + (-2)^2}}
= \frac{-1 - 4}{\sqrt{5} \cdot \sqrt{5}}
= \frac{-5}{5} = -1
$$  

---

#### 6. **Euclidean Distance**

The **Euclidean distance** measures the **straight-line distance** between two vectors (points) in space:  
$$
\text{distance}(\mathbf{a}, \mathbf{b}) = \sqrt{(a_1 - b_1)^2 + (a_2 - b_2)^2 + \dots + (a_n - b_n)^2}
$$

- Unlike **cosine similarity**, which measures **directional similarity**, Euclidean distance measures **absolute difference** in position and doesn't ignore the lengths of the vectors.
- Smaller distance means more similar in both value **and** direction.

Suppose we have two points (or feature vectors):  

$$
\mathbf{a} = \begin{bmatrix} 1 \\ 2 \end{bmatrix} \quad
\mathbf{b} = \begin{bmatrix} 4 \\ 6 \end{bmatrix}
$$  

The **Euclidean distance** is the straight-line distance between these two points:  

$$
\text{distance}(\mathbf{a}, \mathbf{b}) =
\sqrt{(4 - 1)^2 + (6 - 2)^2}
= \sqrt{3^2 + 4^2}
= \sqrt{9 + 16}
= \sqrt{25} = 5
$$  

So the points are **5 units apart** in 2-dimensional space.

### **Normalization and Scaling**

In Machine Learning, numbers are not just numeric values — their **scale** (how big or small they are) can completely change how a model behaves.  
That’s why we often **normalize** or **scale** our vectors and features before using them in ML algorithms.

---

### **1. Normalization**

**Goal:** Change a vector so that its **length (magnitude)** becomes **1**, while keeping its **direction the same**.

$$
\mathbf{x}_{\text{norm}} = \frac{\mathbf{x}}{\|\mathbf{x}\|}
$$  

Here:
- $\mathbf{x}$ = your original vector  
- $\|\mathbf{x}\|$ = the **length** (or magnitude) of the vector  

By dividing the vector by its own length, we make it a **unit vector** (length = 1).  

**Why do this?**
- Sometimes we only care about **direction**, not length.  
- Example: When calculating **cosine similarity**, we want to compare **angles** between vectors, not their sizes.  
  - A short arrow pointing northeast should be considered similar to a long arrow pointing northeast — normalization makes that happen.

---

### **2. Feature Scaling**

**Goal:** Put all features on a similar scale so no single feature dominates the learning process.

Imagine you have two features for each of your samples:
- Height (in meters): values between `1.50 – 2.00`
- Income (in dollars): values between `2,000 – 20,000`

If you don’t scale, the model will treat income as **10,000x more important** than height, simply because the numbers are bigger — even if height is just as important for predicting the target.

#### **A) Min–Max Scaling**

Rescales a value to a fixed range, usually **[0, 1]**:

$$
x_{\text{scaled}} = \frac{x - x_{\min}}{x_{\max} - x_{\min}}
$$  

Where:
- $x$ = the original value  
- $x_{\min}$ = the smallest value in the feature vector  
- $x_{\max}$ = the largest value in the feature vector  

This makes:
- $x_{\min} \to 0$  
- $x_{\max} \to 1$  
- Every other value is proportionally squeezed in between.  

#### **B) Standardization (Z-Score Normalization)**

Centers the feature around **0** and gives it a **standard deviation of 1**:

$$
x_{\text{standardized}} = \frac{x - \mu}{\sigma}
$$  

Where:
- $\mu$ = the **mean** (average) of the feature  
- $\sigma$ = the **standard deviation**  

This creates:
- A new distribution with **mean = 0** and **standard deviation = 1**  
- About 68% of the values will lie between –1 and +1 (if data is roughly normal)

#### **In Summary**

- **Normalization:** Makes individual vectors have length = 1.  
  *Used when direction matters (e.g., cosine similarity, some clustering algorithms).*  

- **Scaling:** Makes features comparable by putting them on the same range or distribution.  
  *Used to ensure fair contribution of each feature to the model.*  

If we **don’t** normalize or scale, models can behave unpredictably, converge slower, or even fail to learn.

### **What The Implementation Below Implies**

Each movie is represented as a **vector of numbers** describing how much **horror, action, comedy, romance, and sci-fi** it has.
Cosine similarity compares the **direction** of these vectors rather than their size — meaning it looks at **pattern of features** rather than absolute numbers.

* Let's say the user has recently watched **World War Z**.
* Its vector is `[7, 8, 1, 1, 4]` → meaning: **high horror, very high action, low comedy/romance, some sci-fi**.
* We compare this pattern to all other movies using the cosine similarity.

The movies with the **most similar pattern of features** get the highest scores.

* **Alien** (94%) → Also high in horror & action, with even more sci-fi → almost the same “direction” as *World War Z*.
* **It** and **The Nun** → Both heavy on horror, medium on action → a close match.
* **The Dark Knight** & **Mission Impossible** → Action-heavy, dark themes, just slightly less horror.

These are the kind of movies that “feel” like *World War Z* — **intense, action-packed, dark atmosphere**.

In the meantime ...

The least similar movies had a completely different pattern:

* **The Notebook, Five Feet Apart, Love Actually** → mostly romance, almost no horror/action.
* **Superbad, Step Brothers** → pure comedy, no horror/action.

Their “directions” in feature space point somewhere completely different, so cosine similarity is very low.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image
import requests
from io import BytesIO
from collections import defaultdict
```

```python
# The movie watched recently, pick any movie from the list below
watched_movie = "World War Z"

# Function for calculating cosine similarity
def cos_similarity(a,b):
    return np.dot(a,b)/(np.sqrt(np.sum(a**2))*np.sqrt(np.sum(b**2)))

# Example movies and manually assigned feature scores (1-10)
# Features order: [horror, action, comedy, romantic, sci_fiction]
movies = [
    {"movie": "The Nun", "vector": np.array([9, 4, 1, 1, 3])},
    {"movie": "World War Z", "vector": np.array([7, 8, 1, 1, 4])},
    {"movie": "Alien", "vector": np.array([9, 7, 1, 1, 9])},
    {"movie": "It", "vector": np.array([8, 5, 3, 2, 3])},
    {"movie": "Fast & Furious", "vector": np.array([1, 10, 3, 4, 3])},
    {"movie": "John Wick", "vector": np.array([1, 10, 2, 2, 2])},
    {"movie": "The Dark Knight", "vector": np.array([3, 9, 1, 3, 3])},
    {"movie": "Mission Impossible", "vector": np.array([1, 10, 2, 2, 4])},
    {"movie": "The Mask", "vector": np.array([1, 4, 9, 5, 3])},
    {"movie": "Deadpool", "vector": np.array([2, 9, 9, 3, 4])},
    {"movie": "Superbad", "vector": np.array([1, 2, 10, 4, 1])},
    {"movie": "Step Brothers", "vector": np.array([1, 2, 10, 3, 1])},
    {"movie": "Titanic", "vector": np.array([2, 5, 1, 10, 5])},
    {"movie": "Five Feet Apart", "vector": np.array([1, 2, 2, 9, 2])},
    {"movie": "The Notebook", "vector": np.array([1, 1, 2, 10, 2])},
    {"movie": "Love Actually", "vector": np.array([1, 2, 5, 9, 2])},
    {"movie": "Inception", "vector": np.array([1, 7, 2, 3, 10])},
    {"movie": "Interstellar", "vector": np.array([1, 6, 1, 3, 10])},
    {"movie": "Avatar", "vector": np.array([1, 8, 2, 5, 10])},
    {"movie": "The Matrix", "vector": np.array([2, 9, 2, 2, 10])}
]
# Create DataFrame
df_movies = pd.DataFrame(movies)

# Find the vector for the watched movie
movie_vec = df_movies[df_movies['movie']==watched_movie]['vector'].values[0]

# Calculate cos similarity between the watched movie and all the movies in the df (excluding the watched movie)
similarities = []
for v in df_movies['vector']:
  similarities.append(cos_similarity(movie_vec,v))
df_movies['similarity'] = similarities
df_movies = df_movies[df_movies['movie']!=watched_movie]

# Display top 5 similar movies to the watched movie
recommends = df_movies.nlargest(5,'similarity')
print(f'5 Movie Recommendations After: {watched_movie} → {movie_vec}')
display(recommends)

# Display 5 of the least similar movies to the watched movie
n_recommends = df_movies.nsmallest(5,'similarity')
print("\n5 Movies Not Recommended:")
display(n_recommends)
```

### **Friend Suggestion Algorithms**

In a social media platform that recommends **new friends** to connect with, we can represent each person by a **vector** of preferences and traits, such as:

- **Age** (years)  
- **Sports interest** (1–10)  
- **Cinema interest** (1–10)  
- **Politics interest** (1–10)  
- **Sense of humor** (1–10)

When the system wants to find people who might get along well, it checks **how close they are** to each other in this **5 dimensional** feature space using **Euclidean distance**.

#### **Why Euclidean Distance?**

For this kind of matching, the **absolute difference** in each preference matters:

- If Person A has **humor = 1** and Person B has **humor = 8**, they likely have very different senses of humor — even if they both score high on sports or cinema.
- Euclidean distance **penalizes big gaps** like this — so people with *closer real scores* get matched.
- If we used **cosine similarity** for this problem, it would only compare the *direction* of the vector, ignoring how big the scores are.  
  - Example: Someone who scores low on *everything* can still have a high cosine similarity with someone who scores high on *everything*, because their *ratios* match but in real life, a $[1, 1, 1, 1, 1]$ person is completely different from a $[9, 9, 9, 9, 9]$ person!

So, for friend recommendations:

**Euclidean distance** is better — because here, distance means *real difference*.  
**Closer distance → more similar taste → better match.**

---

### **Effects of Normalization**

Let's see why **normalizing** features is so important in this problem.

Suppose we have three people represented by those 5 features.

Person A (our reference):  
$$
\mathbf{A} = [20, 9, 8, 5, 8]
$$

Person B (very similar, just older):  
$$
\mathbf{B} = [40, 9, 8, 5, 8]
$$

Person C (different hobbies, same age):  
$$
\mathbf{C} = [20, 1, 3, 9, 2]
$$

#### **Euclidean Distance Without Normalization**

Distance between A and B:
$$
d(A,B) = \sqrt{(40-20)^2 + (9-9)^2 + (8-8)^2 + (5-5)^2 + (8-8)^2}
= \sqrt{(20)^2 + 0 + 0 + 0 + 0}
= 20
$$

Distance between A and C:
$$
d(A,C) = \sqrt{(20-20)^2 + (9-1)^2 + (8-3)^2 + (5-9)^2 + (8-2)^2}
= \sqrt{0 + 64 + 25 + 16 + 36}
= \sqrt{141} \approx 11.87
$$

Even though Person B is almost identical to A (only 20 years older), the algorithm thinks **C is a better match** just because C is closer in age, even though C is *very different in all other interests*.

This is **unfair** — age (due to its bigger range) is dominating the distance!

Let's this time scale every feature to **0–1** range before calculating distances.

- Age range on this platform: $18 \leq \text{age} \leq 60$
- Sports, Cinema, Politics, Humor: already $1 \leq x \leq 10$

Normalization formula (min-max scaling):  
$$
x_{\text{scaled}} = \frac{x - x_{\min}}{x_{\max} - x_{\min}}
$$

**Normalized vectors:**

Person A:
$$
\mathbf{A}_{\text{norm}} =
\left[
\frac{20-18}{60-18},\,
\frac{9-1}{9},\,
\frac{8-1}{9},\,
\frac{5-1}{9},\,
\frac{8-1}{9}
\right]
=
[0.048, 0.889, 0.778, 0.444, 0.778]
$$

Person B:
$$
\mathbf{B}_{\text{norm}} =
\left[
\frac{40-18}{60-18},\, 0.889,\, 0.778,\, 0.444,\, 0.778
\right]
=
[0.524, 0.889, 0.778, 0.444, 0.778]
$$

Person C:
$$
\mathbf{C}_{\text{norm}} =
\left[
0.048, \frac{1-1}{9}, \frac{3-1}{9}, \frac{9-1}{9}, \frac{2-1}{9}
\right]
=
[0.048, 0.000, 0.222, 0.889, 0.111]
$$

#### **Distances After Normalization**

Distance between A and B:
$$
d(A_{\text{norm}},B_{\text{norm}})
= \sqrt{(0.524-0.048)^2 + 0 + 0 + 0 + 0}
= 0.476
$$

Distance between A and C:
$$
d(A_{\text{norm}},C_{\text{norm}})
= \sqrt{(0.048-0.048)^2 + (0.889-0)^2 + (0.778-0.222)^2 + (0.444-0.889)^2 + (0.778-0.111)^2}
\approx 1.35
$$

Now the result makes sense: Person B (very similar) is **much closer** than Person C (different in almost everything).  

This is why **normalizing features is crucial** — it ensures that no single feature (like age) unfairly dominates the match.

---

### **Bottom Line**

- **Euclidean Distance → People / Real Measurements**  
  - Absolute differences matter: age, humor, interests.  
  - Example:  
    - Person A: `[20, 9, 8, 5, 8]`  
    - Person B: `[40, 9, 8, 5, 8]` → distance = 0.476 (normalized)  
    - Person C: `[20, 1, 3, 9, 2]` → distance = 1.35  
  - **Closer distance → better match**  

- **Cosine Similarity → Movies / Taste Patterns**  
  - Direction matters, not magnitude: focus on relative preference ratios.  
  - Example:  
    - Movie A: `[1, 9, 10, 2, 7]`  
    - Movie B: `[2, 18, 20, 4, 14]` → cosine similarity = 1  
    - Euclidean distance would incorrectly penalize Movie B because of scale differences  
  - **Same pattern → high similarity**, ignoring absolute ratings  

**Rule of Thumb:**  
- Use **Euclidean distance** when **absolute values matter**.  
- Use **Cosine similarity** when **pattern/relative ratios matter**.

  - Federer, while still close, has a clearly different distance — reflecting that the model is now fairly accounting for *all* traits, not just one or two.

```python
# Features order: [age, sport, cinema, politics, humor]
user = {"name": "user", "vector": np.array([60, 9, 1, 2, 8])}  # Reference user

# Function for calculating Euclidean Distance between two vectors
def eu_distance(a, b):
    return np.sqrt(np.sum((a - b) ** 2))

# List of some celebrities with their traits
people = [
    {"name": "Cristiano Ronaldo", "vector": np.array([39, 10, 3, 2, 5])},
    {"name": "Lionel Messi", "vector": np.array([37, 10, 3, 1, 6])},
    {"name": "LeBron James", "vector": np.array([39, 10, 4, 2, 7])},
    {"name": "Roger Federer", "vector": np.array([42, 9, 2, 1, 8])},
    {"name": "Serena Williams", "vector": np.array([42, 10, 4, 3, 7])},
    {"name": "Elon Musk", "vector": np.array([52, 6, 5, 9, 5])},
    {"name": "Barack Obama", "vector": np.array([61, 4, 5, 10, 8])},
    {"name": "Vladimir Putin", "vector": np.array([71, 5, 2, 10, 2])},
    {"name": "Joe Biden", "vector": np.array([81, 3, 3, 10, 5])},
    {"name": "Donald Trump", "vector": np.array([77, 3, 2, 10, 6])},
    {"name": "Taylor Swift", "vector": np.array([34, 3, 9, 4, 9])},
    {"name": "Ariana Grande", "vector": np.array([30, 2, 10, 3, 8])},
    {"name": "Beyoncé", "vector": np.array([42, 4, 10, 4, 9])},
    {"name": "Ed Sheeran", "vector": np.array([33, 2, 10, 3, 9])},
    {"name": "Rihanna", "vector": np.array([36, 3, 10, 4, 8])},
    {"name": "Keanu Reeves", "vector": np.array([59, 3, 9, 3, 9])},
    {"name": "Leonardo DiCaprio", "vector": np.array([49, 2, 10, 5, 7])},
    {"name": "Tom Cruise", "vector": np.array([61, 4, 9, 5, 6])},
    {"name": "Emma Watson", "vector": np.array([34, 2, 10, 7, 8])},
    {"name": "Jennifer Lawrence", "vector": np.array([33, 2, 10, 5, 9])},
    {"name": "Ryan Reynolds", "vector": np.array([47, 3, 9, 4, 10])},
    {"name": "Will Smith", "vector": np.array([55, 3, 9, 4, 10])},
    {"name": "Chris Hemsworth", "vector": np.array([40, 5, 9, 3, 8])},
    {"name": "Dwayne Johnson", "vector": np.array([52, 8, 7, 3, 9])},
    {"name": "Kim Kardashian", "vector": np.array([43, 1, 7, 2, 6])},
    {"name": "Oprah Winfrey", "vector": np.array([70, 2, 7, 8, 8])},
    {"name": "Ellen DeGeneres", "vector": np.array([66, 1, 8, 4, 10])},
    {"name": "Pope Francis", "vector": np.array([87, 1, 2, 10, 6])},
    {"name": "Malala Yousafzai", "vector": np.array([26, 1, 4, 9, 7])},
    {"name": "Greta Thunberg", "vector": np.array([21, 1, 3, 10, 6])},
    user
]

# Create DataFrame
df_people = pd.DataFrame(people)

# Friend suggestions WITHOUT normalization
user_vec = df_people[df_people['name'] == 'user']['vector'].values[0]
df_people_no_user = df_people[df_people['name'] != 'user'].copy()

# Calculate Euclidean distances
distances = [eu_distance(user_vec, v) for v in df_people_no_user['vector']]
df_people_no_user.loc[:,'distance'] = distances

print("5 Friend suggestions for a 60-year-old person who likes sports and humor, not interested in cinema and politics")
print("\nWITHOUT NORMALIZATION")
display(df_people_no_user.nsmallest(5, 'distance'))

# Min-Max Normalization
vectors = np.vstack(df_people['vector'].values)
min_vals = vectors.min(axis=0)
max_vals = vectors.max(axis=0)

# Apply Min-Max scaling
vectors_scaled = (vectors - min_vals) / (max_vals - min_vals)
df_people['vector_scaled'] = [np.round(vec, 4) for vec in vectors_scaled]

# Get normalized user vector and drop user from comparison
user_vec_scaled = df_people[df_people['name'] == 'user']['vector_scaled'].values[0]
df_people_scaled = df_people[df_people['name'] != 'user'].copy()

# Calculate Euclidean distances on normalized data
distances_scaled = [eu_distance(user_vec_scaled, v) for v in df_people_scaled['vector_scaled']]
df_people_scaled.loc[:,'distance'] = distances_scaled

print("\nAFTER NORMALIZATION:")
display(df_people_scaled.nsmallest(5, 'distance'))
```

### **Words Embedded in Vectors**

**In modern NLP (Natural Language Processing), the ultimate goal is to teach machines to learn, process, and understand words in a way that is similar to how we humans do.**

But here’s the problem:  
Computers cannot *read*, *hear*, or *think* in words — they can only work with **numbers**.  

So, if we want a machine to “understand” language, our first step must be to **convert words into numbers**, but not just *any* numbers. We need something that actually **preserves the words' meaning and relationships**, so that the machine can work with them in a meaningful way.

When we say we want a model that understands language, the simplest thing that comes to mind is that it should at least recognize **basic relationships between words**.  
For example:  
- It should know that **“milk”** is more related to **“cream”** than to **“football.”**  
- It should recognize that the relationship between **“bee” → “honey”** is similar to **“chicken” → “egg.”**  

These might sound trivial (or even funny), but they are actually the **building blocks of language understanding** — even for humans.  
Think back to those early school exercises where we had to **connect words with similar meanings**. Those tasks were designed to train our brains to form connections and understand relationships between words.

So the core question becomes:  
> **How do we give this ability to machines — when all they understand are numbers?**

The answer is **vectors**.  
By representing words as vectors (like in **Word2Vec**), we give machines a way to **learn patterns and relationships between words automatically** — just like we do as humans. (more on it in `Section06` of this course)

In this simple example, we manually assigned each word a **vector** with three made-up features: `masculinity`, `royalty`, and `oldness`. By using just these numbers, we can already see that our vectors **capture meaningful relationships** between words. But remember that in **real Word2Vec**, these features are **not hand-crafted** and they are **not human-interpretable**.  
The model doesn’t know about “royalty” or “masculinity” explicitly — instead, it **learns abstract numerical patterns** that best describe how words relate to one another.  

All that matters is that:
- **Similar words** end up **closer together** in this space.  
- **Word relationships** turn into **vector relationships** (directions and distances).  

This way word vectors make it possible to do some truly amazing things, such as:

- Grouping similar words together automatically  
- Powering chatbots, translators, search engines, and more  

And all of this works with nothing but **vectors, distances, and directions** under the hood!

---

### **What The Implementation Below Implies**

The vectors we embedded each word into are intentionally simple so that we can easily **see and reason about their meaning**.
Take the word **king**, for example:

**`king → [90, 95, 86]`**
This means:

* **90 masculinity** → highly masculine
* **95 royalty** → very royal
* **86 oldness** → represents an older figure

Put together, this paints the perfect picture of a king — an **old royal male**.

Now here’s the cool part: because these words are vectors, we can tweak their values mathematically.
For example, if we take the **king vector** and **lower its masculinity score**, we get a **less masculine version of a king** — which naturally gives us a **queen**.
Keep changing the features, and eventually you end up with a **girl**, someone **not masculine, not royal, and not old**.

This setup makes it obvious why **Euclidean distance** is useful — we can measure how close two words are in meaning, and even perform **vector arithmetic** to discover analogies like:

$$
\text{king} - \text{man} + \text{woman} ≈ \text{queen}
$$

In other words, our words are no longer just words — they are **points in a mathematical space**, which means we can explore and manipulate them just like numbers!

* **King ⇔ Man ≈ Queen ⇔ Woman**  
  Here, the formula

  $$
  v_4 = v_3 - (v_1 - v_2)
  $$

  means we take the “royal masculinity” of **king**, remove the “man” part, and apply it to **queen** — the result is **woman**, which makes perfect sense: *a queen is just a royal woman, the same way a king is a royal man*.

* **Boy ⇔ Prince ≈ Man ⇔ King**  
  The difference between **boy** and **prince** is the royalty aspect — add the same “royalty difference” to **man** and we get **king**. *So this formula discovers that what makes a boy a prince, would turn a man into a king.*

* **Princess ⇔ Girl ≈ Prince ⇔ Boy**  
  A **princess** is just a royal **girl** — removing royalty gives us **girl**. Do the same for **prince**, and we get **boy** — *the non-royal version of prince*.

* **Man ⇔ Boy ≈ Woman ⇔ Girl**  
  The only real difference between **man** and **boy** is *age* (oldness feature). Apply the same transformation to **woman**, and we naturally get **girl** — *the younger version of woman*.

* **King ⇔ Queen ≈ Prince ⇔ Princess**  
  The difference between **king** and **queen** is masculinity — remove it from **prince**, and we get **princess**. *The same difference a queen has from a king.*

```python
# Features order: masculinity, royalty, oldness
words = [
    {"word": "king", "vector": np.array([90, 95, 86])},
    {"word": "queen", "vector": np.array([6, 92, 80])},
    {"word": "prince", "vector": np.array([87, 65, 16])},
    {"word": "princess", "vector": np.array([3, 53, 12])},
    {"word": "man", "vector": np.array([95, 2, 90])},
    {"word": "woman", "vector": np.array([5, 0, 87])},
    {"word": "boy", "vector": np.array([87, 1, 10])},
    {"word": "girl", "vector": np.array([3, 3, 9])}
]
words_df = pd.DataFrame(words)

# --- Function to solve analogy: w1 - w2 ≈ w3 - ? ---
def solve_analogy(w1, w2, w3):
    """
    Finds w4 such that w1 - w2 ≈ w3 - w4
    Example: w1='king', w2='man', w3='queen' -> returns w4 ≈ 'woman'
    """
    v1 = words_df.loc[words_df['word'] == w1, 'vector'].values[0]
    v2 = words_df.loc[words_df['word'] == w2, 'vector'].values[0]
    v3 = words_df.loc[words_df['word'] == w3, 'vector'].values[0]

    # Solve for v4 using vector arithmetic: v4 ≈ v3 - (v1 - v2)
    v4 = v3 - (v1 - v2)

    # Find the word closest to v4
    words_df['distance'] = words_df['vector'].apply(lambda v: eu_distance(v4, v))
    best_match = words_df.nsmallest(1, 'distance')['word'].values[0]

    return best_match

# --- Example relationships to explore ---
analogies = [
    ('king', 'man', 'queen'),      # king - man ≈ queen - ???
    ('boy', 'prince', 'man'),      # boy - prince ≈ man - ???
    ('princess', 'girl', 'prince'),# princess - girl ≈ prince - ???
    ('man', 'boy', 'woman'),       # man - boy ≈ woman - ???
    ('king', 'queen', 'prince')    # king - queen ≈ prince - ???
]

# --- Run and display results ---
for w1, w2, w3 in analogies:
    w4 = solve_analogy(w1, w2, w3)
    print(f"{w1} ⇔ {w2} ≈ {w3} ⇔ {w4}")
```

## **Matrices**

A **matrix** is a 2D array of numbers arranged in **rows and columns**:

$$
\mathbf{A} = \begin{bmatrix}
a_{11} & a_{12} & \dots & a_{1n} \\
a_{21} & a_{22} & \dots & a_{2n} \\
\vdots & \vdots & \ddots & \vdots \\
a_{m1} & a_{m2} & \dots & a_{mn}
\end{bmatrix}
$$

- This is an **$m \times n$** matrix → $m$ rows, $n$ columns.
- Each element is written as $a_{ij}$, where:
  - $i$ = row index
  - $j$ = column index

### **Common Matrix Operations**

#### 1. **Matrix Addition / Subtraction**

Two matrices can be added only if they have the **same shape**:

$$
\begin{bmatrix} 1 & 2 \\ 3 & -4 \end{bmatrix}
+
\begin{bmatrix} 5 & 6 \\ 7 & 8 \end{bmatrix}
=
\begin{bmatrix} 6 & 8 \\ 10 & 4 \end{bmatrix}
$$

#### 2. **Scalar Multiplication**

Multiply every element by a scalar:

$$
3 \cdot
\begin{bmatrix}
1 & 2 \\
3 & 4
\end{bmatrix}
=
\begin{bmatrix}
3 & 6 \\
9 & 12
\end{bmatrix}
$$

#### 3. **Matrix-Vector Multiplication**

Matrix $\mathbf{A}$ of shape $m \times n$ multiplied by vector $\mathbf{x}$ of shape $n \times 1$:

$$
\mathbf{A} \cdot \mathbf{x} = \mathbf{y}
$$

- Result: a new vector of shape $m \times 1$

#### 4. **Matrix-Matrix Multiplication**

You can multiply matrices **only if** the number of columns in the first equals the number of rows in the second:

If:

$$
\mathbf{A} \in \mathbb{R}^{m \times n}, \quad \mathbf{B} \in \mathbb{R}^{n \times p}
$$

Then:

$$
\mathbf{C} = \mathbf{A} \cdot \mathbf{B} \in \mathbb{R}^{m \times p}
$$

Each element of the result:

$$
c_{ij} = \sum_{k=1}^{n} a_{ik} b_{kj}
$$

### **Matrix Applications in ML**

Linear algebra is not just theory — it's what powers almost **every step** of a machine learning pipeline. Below are key places where **vectors and matrices** show up and what to notice when dealing with them:

### 🔹 1. **Datasets Are Matrices**

  - **Rows** = individual **samples**

- Each sample (row) is essentially a **vector** living in a high-dimensional **feature space**.

#### **The Challenge**

We can’t just sum up the answers because not all traits are equally good or bad. Some should **boost** a player’s score, others should **lower** it, and some should affect the score **more** than others:

- **More important traits** should get **larger weights** so they influence the score more strongly.

In other words, we need a **weighted scoring judgment**, the weights can tell the model which answers we value and which ones should count against a player.

This is where matrices shine:

- **Answers Matrix** → shape **(25×6)**

- **Weights Vector** → shape **(6×1)**

- Multiply them:  
  $$
  \underbrace{[25 \times 6]}_{\text{answers}} \times
  \underbrace{[6 \times 1]}_{\text{weights}} =
  \underbrace{[25 \times 1]}_{\text{scores}}
  $$

This is the exact principle behind **linear models and neural networks**:  
- **Inputs** (features) × **Weights** = **Predictions**

And just like that, we can fairly rank everyone and pick the **top students** for our team — a fast, objective decision based on math instead of gut feeling.

- **Selfishness (–3)** → big penalty  
- **Nervousness (–4)** → biggest penalty

###**Convolving Filters on Images**

When a computer looks at a picture, it doesn’t see colors and objects the way we do — it sees **matrices** (grids of numbers). In a **grayscale image**, this is simply a **2D array** where each cell (pixel) has an intensity value from **0 (black)** to **255 (white)** (or 0-1), the numbers between 0-255 would be different shades of gray. The photo becomes pure numbers — and this opens the door to powerful processing.

<div align="center">
  <img src="https://raw.githubusercontent.com/MamoMGD1/ML_101/main/images/S00_C01/lincoln.png" width="600">
</div>

**Image Credit:** [Original Website](https://assets.runemadsen.com/classes/programming-design-systems/pixels/index.html)

Because an image is a matrix, we can process it mathematically using **kernels** and **convolution**. A **kernel (filter)** is a small matrix (like 3×3) that “slides” over the image matrix, performing a mathematical operation to transform it. This is the heart of how we **blur**, **sharpen**, **detect edges**, or apply any filters in an image. (more on it in `Section05` of this course)

---

#### **Example Kernels:**

- **Box Blur:**  
$$
\begin{bmatrix}
1/9 & 1/9 & 1/9 \\
1/9 & 1/9 & 1/9 \\
1/9 & 1/9 & 1/9
\end{bmatrix}
$$
This kernel averages the values of each 3×3 region. The result is a smoothed, softer image because the fine details get “washed out” by the averaging — like spreading paint with a soft brush.

- **Edge Detector:**  
$$
\begin{bmatrix}
-3 & 0 & 3 \\
-3 & 0 & 3 \\
-3 & 0 & 3 \\
\end{bmatrix}
$$
This kernel checks for **sharp changes** in pixel values. When neighboring pixels have big differences, the kernel amplifies that spot — so edges pop out. This is how computers “see” outlines and might then use to differentiate between shapes and objects.

- **Sharpen:**  
$$
\begin{bmatrix}
0 & -1 & 0 \\
-1 & 5 & -1 \\
0 & -1 & 0 \\
\end{bmatrix}
$$
This kernel highlights the center pixel but subtracts surrounding pixels. The effect boosts contrast at edges, making details crisper.

It shows how simple math with **matrices** can completely change an image — detect outlines, smooth noise, or highlight fine details. And while this example uses small filters, real-world **computer vision** applies the same principles at massive scale:  
- Detecting faces in photos  
- Reading handwriting  
- Powering self-driving cars

Behind the scenes, it’s all about **matrices**, **kernels**, and **convolution** — the same building blocks you’re learning now!

This simple demo is your first step toward understanding **Convolutional Neural Networks (CNNs)** — where these ideas grow into entire architectures that can learn to see.

```python
# Function to convolve a kernel (filter) over a matrix
def convolve(kernel, mat, stride=1):
    h, w = mat.shape
    ker_h, ker_w = kernel.shape

    # Calculate output dimensions
    out_h = (h - ker_h) // stride + 1
    out_w = (w - ker_w) // stride + 1

    # Initialize output
    result = np.zeros((out_h, out_w))

    for i in range(out_h):
        for j in range(out_w):
            window = mat[i*stride : i*stride + ker_h, j*stride : j*stride + ker_w]
            result[i, j] = np.sum(window * kernel)

    return result

# Image URL
img_url = "https://raw.githubusercontent.com/MamoMGD1/ML_101/main/images/S00_C01/cat.jpg"

# Load the image from URL and convert to grayscale
response = requests.get(img_url)
img = Image.open(BytesIO(response.content)).convert('L')

# Convert to NumPy array (matrix)
img_matrix = np.array(img)

# Display the grayscale image
plt.imshow(img, cmap='gray')
plt.axis('off')
plt.title(f'Original Grayscale Image {img_matrix.shape}')
plt.show()

print("Pixel matrix example (top-left corner):")
print(img_matrix[:5,:5])  # Show a 5x5 pixel block

box_blur = np.array([[1/9,1/9,1/9],
                     [1/9,1/9,1/9],
                     [1/9,1/9,1/9]])

edge_detector = np.array([[-3,0,3],
                          [-3,0,3],
                          [-3,0,3]])

sharpen = np.array([[0,-1,0],
                    [-1,5,-1],
                    [0,-1,0]])

filtered_img = Image.fromarray(convolve(edge_detector,img_matrix))
plt.imshow(filtered_img, cmap='gray')
plt.axis('off')
plt.title('Filtered Image')
plt.show()
```

###**Channels in RGB Images**

A colored image is not just a flat picture — to a computer, it’s a **3D array** made of **three overlapping matrices**. For example, the image below is in shape **(506, 900, 3)** which means it’s **506 pixels tall**, **900 pixels wide**, and has **3 channels**: **Red**, **Green**, and **Blue** (**RGB**).

<div align="center">
  <img src="https://raw.githubusercontent.com/MamoMGD1/ML_101/main/images/S00_C01/channels.jpg" width="500">
</div>

Each channel is a **2D matrix** filled with numbers from **0 to 255**:
- **0** means that color is completely absent at that pixel.
- **255** means that color is fully present.

When you separate these channels, you can **visualize how much each color contributes** to the final image:
- The **red channel** shows where the red tones are strong (tulips).
- The **green channel** shows the green parts (tulip stems).
- The **blue channel** shows the blue color distribution (sky).

Think of it like making a pizza:
- The dough alone doesn’t look like a pizza, nor do cheese or tomatoes alone.
- Add cheese, then toppings on top of the dough — layer by layer — and it comes together.
- Similarly, when you stack the red, green, and blue matrices, you get the full image.

You can even **combine channels** (like adding multiple toppings). For example, stacking **green + blue** shows how their overlap shapes the final colors. Notice how the parts of image that are red remain dark in this overlap (the tulips' petals).

In image processing and computer vision, understanding this layered structure is crucial. It’s the foundation for manipulating, enhancing, or analyzing images and their features — all thanks to **matrices**.

```python
# Load the image from URL
img_rgb = "https://raw.githubusercontent.com/MamoMGD1/ML_101/main/images/S00_C01/tulips.jpg"
response = requests.get(img_rgb)
img = Image.open(BytesIO(response.content))
img_matrix = np.array(img)

# Create empty canvas
zeros = np.zeros_like(img_matrix[:,:,0])

# Original RGB Image
plt.figure(figsize=(5,4))
plt.imshow(img)
plt.title(f"Original RGB Image {img_matrix.shape}")
plt.axis('off')
plt.show()

# Red channel in RED
plt.figure(figsize=(5,4))
red_img = np.stack([img_matrix[:,:,0], zeros, zeros], axis=2)
plt.imshow(red_img)
plt.title("Red Channel")
plt.axis('off')
plt.show()

# Green channel in GREEN
plt.figure(figsize=(5,4))
green_img = np.stack([zeros, img_matrix[:,:,1], zeros], axis=2)
plt.imshow(green_img)
plt.title("Green Channel")
plt.axis('off')
plt.show()

# Blue channel in BLUE
plt.figure(figsize=(5,4))
blue_img = np.stack([zeros, zeros, img_matrix[:,:,2]], axis=2)
plt.imshow(blue_img)
plt.title("Blue Channel")
plt.axis('off')
plt.show()

# Green + Blue Channel
plt.figure(figsize=(5,4))
plt.imshow(green_img+blue_img)
plt.title("Green + Blue Channel")
plt.axis('off')
plt.show()
```

Now that you know how images break down into **three color matrices (RGB)**, here’s a simple example that shows how we can **classify an image** just by checking **which color dominates**.

For instance:
- A **sea photo** usually has a lot of blue → **blue channel** has the highest sum.
- A **desert photo** often has more reddish or yellowish tones → **red channel** leads.
- A **forest photo** is rich in greens → **green channel** dominates.

In the code below:
- We load some images.
- For each image we **sum** each channel’s pixel values.
- We compare which sum is the biggest → that decides if we label it as **sea**, **desert**, **forest**, or none.

This works because the **color matrices** store how much of each color is present across all pixels — and simple math on these matrices can already reveal something about the picture.

**But Note That:**
- This is a **very basic approach**.
- For example, a green frog photo might be misclassified as a forest.
- Or a forest photo with lots of sky might be labeled as a sea.

Modern **image classification** goes way beyond this:
- **Convolutional Neural Networks (CNNs)** look at **tiny patterns**, edges, shapes, textures — not just color sums.
- They use many **kernels** and layers to learn complex features automatically.

Still, this simple project is a good reminder: **at the heart of it all, images are matrices**, and

> **computer vision = matrices + math**

```python
# Categories: desert, forest, sea, none
# Function to categorize images based on color intensities
def classify_img(img):
    r,g,b = np.array(img)[:,:,0].sum(),np.array(img)[:,:,1].sum(),np.array(img)[:,:,2].sum()
    if r > g and r > b:
        return 'desert'
    elif g > r and g > b:
        return 'forest'
    elif b > r and b > g:
        return 'sea'
    else:
        return None

# Load images
images_url = 'https://raw.githubusercontent.com/MamoMGD1/ML_101/main/images/S00_C01/img_classification/'+np.array(['d0','d1','d2','d3','f0','f1','f2','f3','s0','s1'])+'.jpg'
classes = []
for url in images_url:
    img = Image.open(BytesIO(requests.get(url).content))
    classes.append({'img': img, 'category': classify_img(img)})

grouped = defaultdict(list)
for item in classes:
    grouped[item['category']].append(item['img'])

# Plot
for category, images in grouped.items():
    n = len(images)
    fig, axs = plt.subplots(1, n, figsize=(n * 2, 2.5))  # 1 row, n columns

    if n == 1:
        axs = [axs]  # Ensure axs is iterable

    fig.suptitle(category, fontsize=18)

    for ax, img in zip(axs, images):
        ax.imshow(img)
        ax.axis('off')

    plt.show()
```
