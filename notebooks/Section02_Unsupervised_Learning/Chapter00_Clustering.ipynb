# **CLUSTERING**

Clustering is one of the most common and powerful tasks in **unsupervised learning**. The goal is simple: **group similar data points together** — even when we don’t know any labels in advance.

Imagine you have a big box of mixed objects — clustering is like sorting them into piles based on color, size, or shape **without anyone telling you which pile is correct**. In the real world, clustering helps us find **natural patterns** in customer data, group similar images, detect anomalies, or even compress data.

---

## **Why is Clustering Useful?**

- **Discover Hidden Patterns:** See how data naturally organizes itself.
- **Segment Users or Items:** Find customer groups for targeted marketing.
- **Preprocessing:** Reduce noise or prepare data for other tasks.
- **Anomaly Detection:** Spot outliers that don’t belong to any cluster.

In short, clustering helps us make sense of **raw, unlabeled data** — turning a messy cloud of points into clear, structured groups.

##**K-Means Clustering**

**K-Means** is one of the simplest and most popular clustering algorithms.  
It works in a few intuitive steps:

1️⃣ **Choose K:**  
   Decide how many clusters (**K**) you want to find.

2️⃣ **Initialize Centroids:**  
   Pick **K random points** in your data space — these will be the first **cluster centers** (called *centroids*).

3️⃣ **Assign Points to Clusters:**  
   For each data point, calculate its distance to each centroid. Assign it to the **nearest centroid** — this forms K groups.

4️⃣ **Update Centroids:**  
   For each cluster, **recompute the centroid** by calculating the **mean position** of all points in that cluster.

5️⃣ **Repeat:**  
   Reassign points and update centroids until the clusters don’t change much — this means the algorithm has **converged**.

---

##**The Core Idea**

- **Assignment Step:** “Which centroid is each point closest to?”
- **Update Step:** “Where should the new centroid be, given its assigned points?”

By repeating this process, K-Means minimizes the **distance between points and their cluster centers**, creating tight, well-separated groups.

---
##**Key Things to Remember**

- You must choose a **good value for K** in advance — this is not always obvious and can take trial and error (or using tools like the Elbow Method).

- **K-Means works best** when clusters are **roughly round (spherical), equally sized, and well separated** in space.  
  If your data has clusters with **complex shapes** (like spirals, long strings, or varying densities), K-Means can easily misclassify points.

- **Outliers can confuse the algorithm**, pulling centroids away from the true center of the real clusters.

- **K-Means is a heuristic algorithm** — it tries to find clusters by minimizing the distance from each point to its nearest centroid, but:
  - It **does not guarantee a global optimum** (the absolute best possible arrangement).
  - It’s very common for K-Means to get stuck in a **local minimum** — a solution that’s good, but not perfect.
  - This happens because the final result depends a lot on where the **initial centroids** are placed.

- To help with this, modern implementations (like `sklearn`’s `KMeans`) run the algorithm multiple times with different random initial centroids and pick the best result.
- Another powerful trick — which we’ve applied here — is to **initialize the first centroids as far apart as possible**. This simple idea inspired a popular variant called **`KMeans++`**, which carefully spreads out the initial centroids to reduce the chance of getting stuck in poor local minima and to produce more reliable clusters.

---

##**The Big Idea**

K-Means is simple, fast, and works surprisingly well for many problems — but its **simplicity is also its weakness**:

- It doesn’t adapt to complicated cluster shapes.
- It can’t handle clusters with very different sizes or densities very well.
- It’s sensitive to bad initial guesses.

K-Means is a great first choice for quick clustering — but for trickier problems, you may need more advanced methods (like DBSCAN or Hierarchical Clustering) that can handle weird shapes, noise, and varying densities better.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def generate_data(k=3, random_state=42):
    """
    Generate synthetic cluster data with random means and standard deviations.

    Returns:
    --------
    X : ndarray
        Generated data points of shape (n_samples, 2)
    """
    np.random.seed(random_state)
    X = []

    # Generate k clusters with random properties
    for i in range(k):
        # Random mean position for this cluster
        means = np.random.randint(0, 10*k, 2)

        # Random standard deviations (spread) for this cluster
        stds = np.random.random(2) * k + 1

        # Random number of points in this cluster
        n_per_cluster = np.random.randint(10, 300)

        # Generate points along X0 dimension
        X0 = np.random.normal(means[0], stds[0], n_per_cluster)

        # Generate points along X1 dimension
        X1 = np.random.normal(means[1], stds[1], n_per_cluster)

        # Stack into 2D points
        cluster = np.column_stack((X0, X1))
        X.append(cluster)

    # Combine all clusters into single array
    X = np.vstack(X)
    return X

# Generate data with 7 clusters
X = generate_data(7, random_state=42)

# Create visualization
plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], s=50, alpha=0.6, linewidth=0.5)
plt.title('Synthetic Cluster Data Visualization', fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Feature X₀', fontsize=13)
plt.ylabel('Feature X₁', fontsize=13)
plt.grid(True, alpha=0.3, linestyle='--', linewidth=0.5)

# Add subtle background
plt.gca().set_facecolor('#f8f9fa')
plt.gcf().patch.set_facecolor('white')
plt.show()
```

```python
class KMeans:
    """
    KMeans clustering algorithm with visualization capabilities.
    Uses smart initialization (farthest point first) and real data points as centroids.
    """

    def __init__(self, n_clusters=3, max_iter=10, tolerance=1e-4, random_state=42):
        """
        Parameters:
        -----------
        n_clusters : int
            Number of clusters to form
        max_iter : int
            Maximum number of iterations
        tolerance : float
            Convergence threshold for centroid movement
        random_state : int
            Random seed for reproducibility
        """
        self.n_clusters = n_clusters
        self.max_iter = max_iter
        self.tolerance = tolerance
        self.random_state = random_state
        self.centroids = None
        self.labels_ = None
        self.inertia_ = None

    def _initialize_centroids(self, X):
        """
        Initialize centroids using farthest point first strategy.
        First centroid is random, subsequent ones are chosen to be far from existing ones.
        """
        rng = np.random.RandomState(self.random_state)
        centroids = []

        # Pick first centroid randomly
        first_idx = rng.choice(X.shape[0])
        centroids.append(X[first_idx])

        # Pick remaining centroids by maximizing distance to existing ones
        for _ in range(1, self.n_clusters):
            # Compute minimum distance from each point to existing centroids
            distances = np.array([
                np.min([np.linalg.norm(x - c) for c in centroids])
                for x in X
            ])
            # Pick the point with the maximum minimum distance
            next_idx = np.argmax(distances)
            centroids.append(X[next_idx])

        return np.array(centroids)

    def fit(self, X, plot=True):
        """
        Fit KMeans clustering to data X.

        Parameters:
        -----------
        X : ndarray of shape (n_samples, n_features)
            Training data
        plot : bool
            Whether to visualize the clustering process
        """
        # Initialize centroids using smart initialization
        self.centroids = self._initialize_centroids(X)

        for iteration in range(self.max_iter):
            # Assign each point to nearest centroid
            distances = np.linalg.norm(X[:, np.newaxis] - self.centroids, axis=2)
            self.labels_ = np.argmin(distances, axis=1)

            # Save old centroids to check convergence
            old_centroids = self.centroids.copy()

            # Update centroids to nearest real data point in each cluster
            for k in range(self.n_clusters):
                points_in_cluster = X[self.labels_ == k]
                if len(points_in_cluster) > 0:
                    # Find the cluster mean
                    mean = points_in_cluster.mean(axis=0)
                    # Find the real point closest to this mean
                    distances_to_mean = np.linalg.norm(points_in_cluster - mean, axis=1)
                    closest_idx = np.argmin(distances_to_mean)
                    self.centroids[k] = points_in_cluster[closest_idx]

            # Plot current iteration
            if plot:
                self._plot_clusters(X, iteration)

            # Check for convergence
            centroid_shifts = np.linalg.norm(self.centroids - old_centroids, axis=1)

            if np.all(centroid_shifts < self.tolerance):
                print(f'Converged in iteration {iteration + 1}')
                break

        # Calculate final inertia (sum of squared distances)
        self.inertia_ = self._calculate_inertia(X)

    def _calculate_inertia(self, X):
        """Calculate within-cluster sum of squared distances."""
        distances = np.linalg.norm(X - self.centroids[self.labels_], axis=1)
        return np.sum(distances ** 2)

    def predict(self, X):
        """Predict cluster labels for new data."""
        distances = np.linalg.norm(X[:, np.newaxis] - self.centroids, axis=2)
        return np.argmin(distances, axis=1)

    def _plot_clusters(self, X, iteration):
        """Plot the current state of clustering."""
        plt.figure(figsize=(6, 4))
        colors = ['red', 'green', 'blue', 'orange', 'purple', 'cyan', 'magenta', 'yellow', 'black', 'pink']

        # Plot each cluster
        for k in range(self.n_clusters):
            points = X[self.labels_ == k]
            plt.scatter(points[:, 0], points[:, 1], s=30, color=colors[k % len(colors)], label=f'Cluster {k+1}')
            plt.scatter(self.centroids[k, 0], self.centroids[k, 1], marker='X', s=200,
                       color=colors[k % len(colors)], edgecolor='black')

        plt.title(f'Iteration {iteration+1}' if iteration+1 else 'K-Means Clusters')
        plt.legend()
        plt.grid(alpha=0.3)
        plt.show()

# Apply on generated X values
m = KMeans(3)
m.fit(X)
```

###**Choosing The Right K**

One of the most important questions when using K-Means is: **How many clusters (K) should we use?**

If you choose **too few clusters**, you might force very different points into the same group, missing important patterns.  
If you choose **too many clusters**, you might split natural groups into meaningless micro-groups — making your model overcomplicated and less useful.

So, we need a way to find a **good balance** between *too simple* and *too complex*.

---

###**The Elbow Method**

A simple and popular technique is the **Elbow Method** (we'd already covered for KNN).

Here’s how it works:

1️⃣ Run K-Means with different values of **K** (for example, from 1 to 10).  

2️⃣ For each K, calculate the **inertia** — this is the *sum of squared distances* between each point and its assigned centroid.  
   - Lower inertia means tighter, more compact clusters.

3️⃣ Plot **inertia vs. K**.

You’ll notice that inertia **always decreases** as K increases — more clusters always means each point can be closer to a centroid.  
But at some point, adding extra clusters provides **only tiny improvements**.

---

###**The “Elbow” Point**

In your plot, look for the point where the inertia stops dropping sharply and starts to level off — like an **elbow** in a bent arm.

✅ **Before the elbow:** Adding more clusters makes a big difference.  
✅ **After the elbow:** More clusters add complexity but little benefit.

The **elbow point** is usually a good estimate for the **optimal K** — simple but effective!

---
###**Don't Forget That:**

- **The Elbow Method is not perfect** — sometimes the bend is clear, sometimes it’s subjective.
- Other techniques (like Silhouette Score) can help too.
- But the Elbow Method is a quick, intuitive way to guide your choice — and it’s a great first check!

```python
# Calculate inertia for different k values
inertia = []
for i in range(1, 11):
    # Fit KMeans model with i clusters
    model = KMeans(i)
    model.fit(X, plot=False)
    inertia.append(model.inertia_)

# Create enhanced elbow plot
plt.figure(figsize=(8, 6))
plt.plot(range(1, 11), inertia, marker='o', linewidth=2.5,
         markersize=10, color='#2E86AB', markerfacecolor='#A23B72',
         markeredgecolor='white', markeredgewidth=2)

# Styling
plt.title('Elbow Method: Inertia For Different K Values', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Number of Clusters (k)', fontsize=12)
plt.ylabel('Inertia (Within-Cluster Sum of Squares)', fontsize=12)
plt.grid(True, alpha=0.3, linestyle='--', linewidth=0.8)
plt.gca().set_facecolor('#f8f9fa')
plt.show()
```

**Result:**

It can be seen that increasing the **K** value from 1 to 2, and from 2 to 3, provides significant improvements — this is clear from the steep drop in the loss. However, once **K** reaches 4 or higher, the improvements become marginal. The slope flattens out, showing that adding more clusters beyond this point does not bring much additional benefit.

- Some goods might break simple rules: for example, mid-cost items with unusual profit spikes. A fixed threshold would miss this hidden group.

```python
def generate_business_data(n_samples=300, random_state=42):
    np.random.seed(random_state)

    # Group 1: Low cost, high price (high margin goods)
    n1 = n_samples // 3
    cost1 = np.random.normal(loc=20, scale=5, size=n1)  # low cost
    price1 = np.random.normal(loc=80, scale=15, size=n1)  # high price

    # Group 2: High cost, low price (low margin or loss leaders)
    n2 = n_samples // 3
    cost2 = np.random.normal(loc=70, scale=10, size=n2)  # high cost
    price2 = np.random.normal(loc=50, scale=8, size=n2)  # moderate price

    # Group 3: Medium cost, medium price (average goods)
    n3 = n_samples - n1 - n2
    cost3 = np.random.normal(loc=45, scale=7, size=n3)
    price3 = np.random.normal(loc=55, scale=7, size=n3)

    # Stack into a single (n_samples, 2) array
    cost = np.concatenate([cost1, cost2, cost3])
    profit = np.concatenate([price1, price2, price3])

    X = np.column_stack((cost, profit))
    return X

n=300
business_data = generate_business_data(n)

plt.figure(figsize=(6,4))
plt.scatter(business_data[:,0],business_data[:,1])
plt.grid()
plt.xlabel('Cost (USD)')
plt.ylabel('Profit (USD)')
plt.title(f'Profits vs Costs of {n} Goods')
plt.show()
```

In short, K-Means lets Habib see **what’s really happening** inside his messy storage room, with no labels of **“Best-seller”** or **“Low profit”**, far better than guessing or drawing clumsy lines by hand.

##**DBSCAN**

We saw some limitations in K-Means clustering algorithm. In places where K-Means starts to struggle, **DBSCAN** often shines.  
Unlike K-Means, which cuts space with straight lines and assumes round, equally-sized clusters, **DBSCAN** finds clusters by looking for **dense regions** — no matter what shape they are.

---

###**How DBSCAN Works**

DBSCAN stands for **Density-Based Spatial Clustering of Applications with Noise** (yes, the name is huge but the idea is simple).

**Core ideas:**
- A cluster is a region where points are **densely packed** together.
- Points in sparse regions (between clusters) are labeled **noise** — no forced assignment!
- It uses two main rules:
  - **eps (ε)**: the maximum distance for two points to be neighbors.
  - **min_samples**: the minimum number of neighbors to form a dense region.

**How it finds clusters:**

1️⃣ Pick any point that hasn’t been visited.  
2️⃣ If it has enough neighbors within **eps**, it’s a **core point** — it starts a new cluster.  
3️⃣ All directly reachable neighbors get added to this cluster.  
4️⃣ Keep expanding the cluster by visiting neighbors’ neighbors — if they are also dense enough, they grow the cluster further.  
5️⃣ If a point isn’t part of any dense region, it’s marked as **noise** (`-1`).

---

###**Why DBSCAN is Powerful**

  <img src="https://raw.githubusercontent.com/MamoMGD1/ML_101/main/images/S02_C00/dbscan.png" width="600">
</div>

**Image Credit:** [Original Website](https://mrinalyadav7.medium.com/dbscan-algorithm-c894701306d5)

DBSCAN **fixes several major weaknesses of K-Means:**

- **Odd shapes?**  
  No problem — DBSCAN handles clusters of any shape: circles, spirals, moons… K-Means can’t.

- **Outliers?**  
  K-Means *forces* every point into a cluster — DBSCAN can label isolated points as noise, keeping clusters clean.

- **No need to choose K!**  
  DBSCAN doesn’t require you to guess the number of clusters in advance. Instead, you set `eps` and `min_samples` — clusters emerge naturally.

---

###**What’s the Catch?**

- DBSCAN can struggle when clusters have **very different densities** — one dense, one sparse might confuse it.
- Picking a good `eps` can take a bit of trial and error — but a simple plot often helps.
- Not ideal for extremely high-dimensional data.

---

###**Key Takeaway**

If your data has **weird shapes**, **natural noise**, or you don’t know how many clusters there should be — DBSCAN is your best friend.  
It shows that clustering isn’t just about drawing circles — it’s about letting **structure emerge** from the data, naturally.


```python
class DBSCAN:
    def __init__(self, eps=0.5, min_samples=5):
        self.eps = eps
        self.min_samples = min_samples
        self.labels_ = None

    def fit_predict(self, X):
        """
        Perform DBSCAN clustering and return cluster labels.

        Parameters:
        -----------
        X : ndarray of shape (n_samples, n_features)
            Training data

        Returns:
        --------
        labels : ndarray of shape (n_samples,)
            Cluster labels for each point. -1 indicates noise.
        """
        n_samples = X.shape[0]

        # Initialize all points as unvisited (label = 0 means unvisited)
        self.labels_ = np.zeros(n_samples, dtype=int)

        # Current cluster ID (starts at 1, -1 is reserved for noise)
        cluster_id = 0

        # Visit each point
        for point_idx in range(n_samples):
            # Skip if already processed
            if self.labels_[point_idx] != 0:
                continue

            # Find all neighbors within eps distance
            neighbors = self._get_neighbors(X, point_idx)

            # If not enough neighbors, mark as noise (for now)
            if len(neighbors) < self.min_samples:
                self.labels_[point_idx] = -1
            else:
                # This is a core point - start a new cluster
                cluster_id += 1
                self._expand_cluster(X, point_idx, neighbors, cluster_id)

        return self.labels_

    def _get_neighbors(self, X, point_idx):
        # Calculate distances from point to all other points
        distances = np.linalg.norm(X - X[point_idx], axis=1)

        # Return indices of points within eps distance
        neighbors = np.where(distances <= self.eps)[0]
        return neighbors

    def _expand_cluster(self, X, point_idx, neighbors, cluster_id):
        """
        Expand cluster by adding all density-reachable points.

        This is the core of DBSCAN - it grows the cluster by:
        1. Adding the current point to the cluster
        2. Checking each neighbor
        3. If a neighbor is also a core point, add its neighbors too

        Parameters:
        -----------
        X : ndarray
            Dataset
        point_idx : int
            Index of the starting core point
        neighbors : ndarray
            Indices of neighbors of the core point
        cluster_id : int
            Current cluster ID to assign
        """
        # Add the core point to the cluster
        self.labels_[point_idx] = cluster_id

        # Use a list to track points to process (neighbors might grow)
        i = 0
        while i < len(neighbors):
            neighbor_idx = neighbors[i]

            # If this point was marked as noise, it's now a border point
            if self.labels_[neighbor_idx] == -1:
                self.labels_[neighbor_idx] = cluster_id

            # If this point hasn't been visited yet
            elif self.labels_[neighbor_idx] == 0:
                # Add it to the cluster
                self.labels_[neighbor_idx] = cluster_id

                # Find its neighbors
                neighbor_neighbors = self._get_neighbors(X, neighbor_idx)

                # If it's also a core point, add its neighbors to expand further
                if len(neighbor_neighbors) >= self.min_samples:
                    neighbors = np.concatenate([neighbors, neighbor_neighbors])

            i += 1
```

```python
from sklearn.datasets import make_moons, make_circles, make_blobs

# DATASET 1: Moons
X_moons, _ = make_moons(n_samples=300, noise=0.05, random_state=0)
dbscan_moons = DBSCAN(eps=0.15, min_samples=5)
labels_moons = dbscan_moons.fit_predict(X_moons)

# DATASET 2: Circles
X_circles, _ = make_circles(n_samples=450, factor=0.5, noise=0.05, random_state=0)
dbscan_circles = DBSCAN(eps=0.13, min_samples=5)
labels_circles = dbscan_circles.fit_predict(X_circles)

# DATASET 3: Blobs + Noise
X_blobs, _ = make_blobs(n_samples=250, centers=3, cluster_std=0.4, random_state=0)
noise = np.random.uniform(low=-3, high=6, size=(20, 2))
X_blobs = np.vstack([X_blobs, noise])
dbscan_blobs = DBSCAN(eps=0.3, min_samples=5)
labels_blobs = dbscan_blobs.fit_predict(X_blobs)

# DATASET 4: Custom
dbscan_custom = DBSCAN(eps=5, min_samples=3)
labels_custom = dbscan_custom.fit_predict(X)


# ===== PLOTTING FUNCTION =====

def plot_dbscan(X, labels, title):
    plt.figure(figsize=(5, 4))
    unique_labels = set(labels)
    colors = plt.get_cmap('tab10', len(unique_labels))

    for label in unique_labels:
        # Use black for noise points
        color = 'k' if label == -1 else colors(label)
        class_mask = (labels == label)
        xy = X[class_mask]
        plt.scatter(xy[:, 0], xy[:, 1], s=50, c=[color],
                   label=f'Cluster {label}' if label != -1 else 'Noise')

    plt.title(title)
    plt.grid(alpha=0.3)
    plt.legend()
    plt.show()


# ===== PLOT ALL DATASETS =====
plot_dbscan(X_moons, labels_moons, 'DBSCAN Clustering - Moons')
plot_dbscan(X_circles, labels_circles, 'DBSCAN Clustering - Circles')
plot_dbscan(X_blobs, labels_blobs, 'DBSCAN Clustering - Blobs with Noise')
plot_dbscan(X, labels_custom, 'DBSCAN Clustering - Custom Dataset')
```

###**Fire Detection With DBSCAN**

One of the most powerful strengths of **unsupervised learning** is its ability to uncover unusual or unexpected patterns — **without needing explicit labels**. A great example of this is using **DBSCAN** for anomaly detection.

#### **How Does It Work?**

At its core, DBSCAN looks for **clusters of similar points** and labels anything that doesn’t fit as **noise**. This means that if you can describe your data in a way where “normal” looks like one or more dense clusters, anything that falls *far away* can be treated as an anomaly.

In our example, each image is summarized by the **average values of its RGB color channels** — so each frame becomes a single point in 3D color space (X axis is gonna show its Red channel's average value, Y axis for Green channel and Z axis for Blue channel). DBSCAN then tries to find **dense groups** of similar color distributions. If a new image is radically different — say, a sudden surge of red, or an unusual brightness — it won’t belong to the main cluster and will be flagged as an **outlier**.

The nice part? This does not just apply to **fire**. If your camera sees **any unexpected scene change** — a person walking into a forbidden zone, a door left open, an unusual object — the mean color can shift enough to break from the normal cluster. With the right sensitivity, DBSCAN will detect it.

---

#### **Tuning Sensitivity**

The **`eps` parameter** controls *how close* points need to be to be considered neighbors:
- A **low `eps` (e.g. 1)** makes DBSCAN very sensitive — only frames almost identical to the normal cluster will pass as normal.
- A **higher `eps`** widens the tolerance — useful if you want to allow some natural variation.

Fine-tuning `eps` and `min_samples` helps balance **false positives** (normal changes flagged as anomalies) and **false negatives** (real anomalies slipping through).

The true power of DBSCAN for anomaly detection is that it’s **domain-agnostic**:  
- **Images** → detect unusual scenes, intrusions, or equipment failures.  
- **Sound** → detect strange vibrations or abnormal engine noise by clustering frequency patterns.  
- **Sensor streams** → detect outlier temperature, vibration, or pressure readings in factories.  
- **Log files or events** → catch rare or suspicious events that don’t match normal usage.

Any time your data can be described in a space where “normal” is dense and “weird” is isolated, DBSCAN is a candidate.

---

### **Strengths & Limitations**

✅ **Why it’s great:**
- **No labels needed.**
- Can detect *any* unusual pattern, not just predefined ones.
- Works on any feature space — colors, sounds, stats.

⚠️ **Why it’s not enough alone:**
- Subtle, slow anomalies can *creep* and stay within the normal cluster.
- Local changes may be hidden if you only use *global* features (like a mean).
- It can be fooled by natural variations if `eps` is not tuned carefully.

**In real systems**, DBSCAN is often part of a **bigger pipeline**:
- Use it for a quick filter — to flag possible outliers.
- Follow up with more advanced models: local patch analysis, motion detection, heat sensors, or trained classifiers.
- Combine multiple sensors (visual + thermal + sound) for robust detection.

```python
import pandas as pd
from collections import defaultdict
from PIL import Image
import requests
from io import BytesIO
import math

images_url = 'https://raw.githubusercontent.com/MamoMGD1/ML_101/main/images/S02_C00/fire/'+np.array(['normal','normal','bright','fire','dark','fire2','fire3','normal','dark','bright'])+'.png'

# Load images and compute channel means
rgb_means = []
images = []

for url in images_url:
    img = Image.open(BytesIO(requests.get(url).content)).convert('RGB')
    img_array = np.array(img)
    mean_rgb = img_array.mean(axis=(0, 1))  # (H, W, 3) → (3,)
    rgb_means.append(mean_rgb)
    images.append(img)

rgb_means = np.array(rgb_means)  # (N, 3)

# Apply DBSCAN
fire_db = DBSCAN(eps=1, min_samples=2)
labels = fire_db.fit_predict(rgb_means)

# Build classes list
classes = []
for img, label in zip(images, labels):
    category = "Abnormal" if label == -1 else "Normal"
    classes.append({'img': img, 'category': category})

# Group for plotting
grouped = defaultdict(list)
for item in classes:
    grouped[item['category']].append(item['img'])

# Plot
n = len(classes)
n_cols = math.ceil(n / 2)  # How many images per row
n_rows = 2

fig, axs = plt.subplots(n_rows, n_cols, figsize=(n_cols * 3.5, n_rows * 5))

# Flatten axes in case it's a 2D array
axs = axs.flatten()

for i, (ax, item) in enumerate(zip(axs, classes)):
    img = item['img']
    category = item['category']

    # Show image
    ax.imshow(img)
    ax.axis('off')

    # Title on top
    ax.set_title(f"Frame {i+1}", fontsize=15, pad=10)

    # Label below
    color = 'green' if category == "Normal" else 'red'
    fontweight = 'bold' if category == "Normal" else 'black'
    ax.text(
        0.5, -0.1, category,
        color=color, fontsize=20, fontweight=fontweight,
        ha='center', va='center',
        transform=ax.transAxes
    )

# If there are empty subplots, hide them
for j in range(i + 1, len(axs)):
    axs[j].axis('off')

plt.tight_layout()
plt.show()
```

### **Limitations**

As you can see, **DBSCAN** is not always the perfect choice either.  
For example, the custom dataset we generated for **K-Means** — nice, tight, well-separated blobs — works beautifully with centroid-based clustering but **doesn’t fit DBSCAN’s strengths**. Because DBSCAN depends on **density**. When clusters are **very close together** or **densities vary a lot**, DBSCAN can:
- Merge nearby clusters into one big blob,
- Or break a single cluster into scattered parts,
- Or label too many points as noise.

---

###**The Right Tool for the Right Shape**

- **K-Means** is great for **compact, round, balanced clusters** — but struggles with unusual shapes or noise.
- **DBSCAN** is fantastic for **arbitrary shapes and noise** — but needs clear density gaps to work well.
