# **REINFORCEMENT LEARNING**

So far, our machine learning journey has taught us how to **observe patterns** (supervised learning) and **discover structure** in raw data (unsupervised learning). These techniques helped us predict outcomes and find hidden relationships. But now, it’s time to give our models something new:

> **The power to act.**

Reinforcement Learning (RL) is where machines stop being passive learners and start becoming decision-makers like a child learning to walk, or a robot learning to play chess. Here, the model is no longer handed perfect answers or tidy datasets. Instead, it must **learn by doing**, **learn by trying**, and most importantly — **learn by failing.**

This is the third pillar of Machine Learning, and it’s different. Very different.
It’s **not about learning from data**, it’s about learning from **experience.**

---

## **The Language of Reinforcement Learning**

The learning process in RL begins when an **agent** enters an **environment** without having a clue about what it should do or even what it's trying to achieve. It just knows it can take some **actions**, and every action leads it to a new **state** in the environment — along with a **reward** that says, "Good job!" or "Big mistake."

Over time, through countless decisions and feedback, it starts figuring out the puzzle of what’s beneficial and what’s not. It begins learning *which actions* in *which situations* tend to bring better results.

This is not too different from how we grow up. Think about a baby:

They (as **agents**) come into the world with little knowledge about anything. The **environment** is this strange, colorful, noisy world filled with people and objects. Early **actions** like crying, crawling, or reaching for a toy may sometimes result in a **reward** — food, comfort, joy.

Over time, the baby figures out a kind of **policy** — a strategy: “If I cry, someone comes.” Later, they learn more complex behaviors: “If I study, I get praise or good grades.” “If I exercise, I feel better.” “If I work hard, I might get paid.”

That’s Reinforcement Learning at its core: doing things, seeing what happens, and slowly discovering how to live well in the environment you’re placed in.

Let’s break it down into simple concepts:

- **Agent**: The decision-maker. Think of it like the player in a video game.
- **Environment**: The world in which the agent lives. The rules, the terrain, everything that reacts to the agent.
- **State**: A snapshot of the current situation. It’s what the agent can “see” or “know” about the world right now.
- **Action**: A choice the agent makes to interact with the environment.
- **Reward**: A signal sent by the environment to say how good or bad the action was.
- **Policy**: The agent’s strategy — a rule or function that decides what action to take in each state.
- **Episode**: A full sequence of interactions, from beginning to end (like one round of a game).
- **Value**: The long-term benefit of being in a certain state or taking a certain action.

# **MONTE CARLO ALGORITHM**

The name *Monte Carlo* might sound like a fancy Italian dessert — but it’s actually inspired by the famous casino in Monaco. Why? Because the essence of this method is **random sampling** and **probabilistic learning** — just like placing bets in a game of chance.

In **Reinforcement Learning**, Monte Carlo (MC) methods are a class of algorithms that learn **from experience**, but specifically from **complete episodes** — full sequences of states, actions, and rewards from beginning to end.

---

## **What’s the Big Idea?**

Monte Carlo methods answer this core question:

> “If I keep trying different things and observing the final result, can I eventually figure out what works best?”

And the answer is yes — with **enough experience**, Monte Carlo methods can learn good decision strategies by averaging the rewards received **after visiting a certain state or taking a certain action.**

Let’s say we’re trying to teach an agent how to play a simple game. We let it play thousands of full games (called *episodes*), and at the end of each game, we look back and say:

> “What actions did we take? What situations did we visit? How well did that game go?”

We then assign credit (or blame) to each move based on how the **entire game** turned out.

## **Decision Making**

Here’s the workflow:
1. The agent plays many **episodes** (full games).
2. After each episode, it looks at the total reward it received.
3. It goes back through each **state-action pair** it visited.
4. It updates an average score for each pair based on that total reward.

This average becomes the **estimated value** of that decision in that situation.

---

### **Example 1: A Tennis-Playing Agent**

Let’s imagine you’re building a robot agent to learn tennis.

You don’t teach it rules. Instead, you let it play thousands of full matches, collecting observations like:

- State: *The ball is approaching me diagonally.*
- Action: *Swing the racket with medium force.*
- Result: *Opponent missed. I scored.*

Another episode might look like:

- State: *Ball is approaching slowly.*
- Action: *Turn around and run away.*
- Result: *I lost the point.*

Over time, the agent starts to learn:
- Taking the action **"swing at the ball"** tends to result in **positive outcomes**.
- Taking the action **"run away from the ball"** almost always leads to **negative rewards**.

And even more specifically:
- The **best** outcomes came from the situations where it hit the ball just right and the **opponent couldn’t return** it.

So it increases the estimated value of:  
→ *(Approaching ball → swing with control)*  
and decreases the value of:  
→ *(Approaching ball → run away)*

The more the agent plays, the better its estimates get. Its decisions become less random, and more based on **proven experience**.

---

### **Example 2: Learning to Drive**

A self-driving car agent may simulate 1000 full trips across a city.

- It finds that when it **slows down before sharp turns**, it gets a good reward (no crashes, smoother ride).
- When it **rushes through intersections**, the episodes often end with a crash (bad reward).

So it assigns **high value** to cautious behavior in certain contexts.

---

### **Example 3: Fast-Food Worker Agent**

Imagine training an AI assistant in a fast-food kitchen:
- It tries **different sequences** of preparing a burger.
- In some episodes, customers complain (low reward).
- In others, the burger is served fast and perfectly (high reward).

After 5000 episodes, it learns:
> “Grilling the meat before adding the lettuce gives consistently better results.”

Even though it gets no reward until the end, it still figures out what steps **led to success.**

Monte Carlo teaches us this:  
**Experience, repeated many times, will tell us what works the best and what to prevent — even if we can’t judge it moment by moment.**

## **Limitations of Monte Carlo**

Monte Carlo isn’t perfect. It has some key limitations:

* **Requires Full Episodes:**
You only update your knowledge **after the whole episode ends.** This means it's slow to learn in very long or never-ending environments (like driving a car continuously, or managing a power grid).

* **Slow to Converge:**
Since rewards are noisy, you need **lots of episodes** to get accurate estimates.

* **High Variance:**
The randomness of early actions can drastically affect the final outcome. This makes it harder to tell which move was really responsible for the success or failure.

* **Credit Assignment Problem:**
In long sequences, it’s hard to know which actions *really* mattered most. Was it the early good move that helped us win, or a lucky choice later on?

---

## **When Should You Use It?**

Monte Carlo is powerful when:
- You can simulate or observe many episodes.
- The environment is **episodic** (it has a clear start and end).
- You want to learn from **real results**, not just predictions.
- You’re okay with slower learning in exchange for **true long-term understanding.**

## **Implementation on a Game**

We create a simple yet fun game:

> There are 🍎🍎🍎🍎... apples placed in a row.  
> Players take turns to **remove 1, 2, or 3 apples**.  
> The one who takes the **last apple** loses the game.  
> If you leave your opponent with no choice but to take the last one, you win.

It’s a bit like the game of **Nim**, with 🍎 apples instead of sticks.

### **The Game Environment**

Before an agent can learn to play, it needs a **world** to play in.  
That’s exactly what the `Environment` class builds.

* `self.board`: A list of 🍎 emojis representing the current state.
* `self.aiTurn`: Keeps track of whose turn it is.
* `self.winner`: Will hold `'AI'` or `'Human'` when the game ends.

So basically the `Environment` class:

* Provides a **game board** the agent can learn on,
* Defines **rules**, **winning conditions**, and **available actions**,
* Tracks the **current state** of the game and the player.

It's the perfect little world for our learning agent to **try**, **fail**, and eventually... **get really good**.

`````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
`````

`````python
class Environment:
    def __init__(self, board_length=11):
        self.board = ['🍎' for _ in range(board_length)]
        self.aiTurn = True
        self.winner = None

    def show_board(self):
        print(''.join(self.board))

    def apply_action(self, action):
        self.board = self.board[action:]
        self.aiTurn = not self.aiTurn
        if len(self.board) == 0:
            self.winner = 'AI' if self.aiTurn else 'Human'

    def valid_actions(self):
        return list(range(1, min(3, len(self.board)) + 1))
`````

### **The Trainee & The Trainer**

At the beginning, our AI doesn’t know:
- what the goal is,
- what actions are “good” or “bad”,
- or even how the game ends!

It just knows:
> “Here’s the number of apples left. You can take 1, 2, or 3. Decide and see what happens.”

That’s it. There’s no teacher. No manual. No hints. Just trial and error. We then let the agent **play the game over and over again**.  

Each time, it either wins (gets reward `+1`) or loses (gets reward `–1`).

It saves **which moves it took** in which **board state**.  
And eventually learns:
> “Hmm… every time there were 3 apples and I took 2, I tended to win.”

So it increases its confidence in that move.

Bit by bit, it discovers which moves are **strategically powerful**, which moves **provide a state of 3 apples left for me**, and which are **risky or foolish.**

---

### **The Brain of the Agent (So Far)**

The `Player` class is the brain of our agents.  
Each agent has a `records` dictionary — this is its **entire memory**, storing every experience it has had:

```python
self.records = {}  # (state, action): [list of rewards]
````

* The **key** is a pair of `(state, action)`.
* The **value** is a list of results it got when it did that.

So if it saw the board `[🍎,🍎,🍎,🍎,🍎]` (5 apples) and removed 2, then won, it would remember:

```python
{(state=(🍎,🍎,🍎,🍎,🍎), action=2): [1]}
```

If it tries the same move again in a different episode and loses, it becomes:

```python
{(state=(🍎,🍎,🍎,🍎,🍎), action=2): [1, -1]}
```

This way, it slowly builds up **statistical estimates** of how good each move is in every possible board state.

---

### **Why Would This Work?**

The magic of this learning method lies in **repetition and averaging**.

If an agent tries an action **dozens, hundreds, or thousands of times**, the average reward it gets will **start to converge** toward the true value of that action in that situation.

Over time:
- Random noise and luck cancel out.
- The agent starts seeing **the real value** of each move.

For example, imagine the agent has seen:
```python
(state=(🍎,🍎,🍎,🍎,🍎,🍎), action=1): [1, -1, 1, 1, -1, 1]
````

That's a 4/6 win rate → 66.6% success.

With more games, that percentage becomes more meaningful — it smooths out spikes from short-term randomness.

---

### **Why Do We Use a "Cruel" Trainer?**

At first glance, this setup seems upside-down:

* The **AI Trainee** plays completely **randomly**, with no memory or strategy.
* The **Cruel Trainer** watches the AI fail over and over — and **learns from its mistakes**.

But there's deep wisdom in this structure:

1. **Maximal Exploration by Design**  
   A random agent guarantees variety. It plays well sometimes, badly often, and does *everything in between*.
   This feeds the trainer with a **rich dataset of decisions and consequences**, avoiding the trap of narrow learning.

2. **Freezing the Trainee Adds Stability**  
   If both sides were learning, things would spiral into chaos.
   By keeping the AI naive and random, the trainer can **learn patterns from a stable distribution of play**, episode after episode.

3. **Clear Division of Labor**

   * **Trainee = Experience generator**
   * **Trainer = Insight extractor**  
     The AI blindly stumbles through the world, while the Trainer becomes increasingly wise by *watching from the sidelines*.

> Imagine an eager toddler thrown into a game it doesn’t understand. It pushes buttons at random.
> Behind the glass, a seasoned scientist takes notes — watching what works, what fails, and **slowly building a winning strategy**.

### **How Does The Agent Makes Smart Choices?**

Let’s now decode the agent’s thought process during the game.

The method:

```python
def decide(self, board):
    state = tuple(board.board)
```

First, the board (e.g. `[🍎,🍎,🍎,🍎]`) is converted into a **tuple**, so it can be used as a key in a dictionary.

```python
    if self.name == 'AI':
        possible = [
            (k, np.mean(v))
            for k, v in self.records.items()
            if k[0] == state
        ]
```

Now the agent **searches through its memory**:

* Finds all records where the **state** matches the current board,
* Averages their scores (using `np.mean(v)`).

Each option becomes something like:

```python
[((🍎,🍎,🍎,🍎,🍎,🍎), 1), 0.75]  # Taking 2 apples from 5 seems to give 75% win
```
Out of all those options, it **picks the one with the highest average reward** — the best move it has seen **so far**.

```python
        if possible:
            best_action = max(possible, key=lambda x: x[1])[0][1]
            return best_action
```

If it has no data on the current state, it goes with a random choice. That’s pure exploration — trying something new.

And if the agent is a human (i.e. not AI or Trainer), it just asks for input.

`````python
class Player:
    def __init__(self, name):
        self.name = name
        self.records = {}  # (state, action): [list of returns]

    def decide(self, board):
        state = tuple(board.board)

        if self.name == 'AI':
            # Filter all matching state-action pairs
            possible = [(k, np.mean(v)) for k, v in self.records.items() if k[0] == state]

            if possible:
                # Choose the action with the highest average return
                best_action = max(possible, key=lambda x: x[1])[0][1]
                return best_action

            # No data? Pick randomly
            return np.random.choice(board.valid_actions())
        else:
            # If player is a human
            while True:
                move = int(input('Enter your move: '))
                if move in board.valid_actions():
                    break
            return move

    def train(self, episodes):
        for _ in range(episodes):
            p1 = self               # Trainee (will not evolve)
            p2 = Player('trainer')  # Trainer (improves across games)
            board = Environment()
            episode = []        # store (state, action) pairs for AI only

            while not board.winner:
                p = p1 if board.aiTurn else p2
                move = (
                    np.random.choice(board.valid_actions())
                    if p == p1 else p1.decide(board)
                )
                if p == p1:
                    episode.append((tuple(board.board), move))
                board.apply_action(move)

            # At the end of episode, assign final reward
            reward = 100 if board.winner == 'AI' else -100

            # Every-Visit Monte Carlo: update all (s,a) pairs
            for state_action in episode:
                if state_action not in self.records:
                    self.records[state_action] = []
                self.records[state_action].append(reward)
    def plot_records(self):
        df = pd.DataFrame([
            {'apples': len(s), 'action': a, 'score': np.mean(v)}
            for (s, a), v in self.records.items()
        ])
        pivot = df.pivot_table(index='apples', columns='action', values='score')

        plt.figure(figsize=(10,6))
        plt.title("Preferred Actions by Apples Left With Their Winning Rate")
        sns.heatmap(pivot, annot=True, fmt=".2f", cmap='coolwarm', cbar_kws={'label': 'Average Score'})
        plt.xlabel("Action Taken (apples removed)")
        plt.ylabel("Apples Left on Board")
        plt.gca().invert_yaxis()
        plt.show()
`````

### **The Winning Policy**

After playing **10,000 games**, something remarkable happens — even though no one told the agent how to win, it starts making **very intelligent moves**.

Let’s break down *why* this learning process works, and what the resulting heatmap tells us.

**We could notice how:**

1. **Exploration Creates Understanding:**
The agent tried **every kind of move** — smart, stupid, and risky — simply by playing games. Over time, it saw what **outcomes** each move led to.

2. **Experience Becomes Knowledge:**
For every board state (e.g., 5 apples left), it recorded the **average reward** of each action based on past experiences.
Thanks to the **Law of Large Numbers** and the **Central Limit Theorem**:
   > With enough games, these averages **converge** to the **true expected outcome** of each move.

3. **Patterns Begin to Emerge:**
From random chaos, the agent starts to **see the structure of the game**.
It learns which moves lead to **victory** and which are almost always **losses**.

---

Let’s now interpret the strategic insight behind the heatmap below:

#### **1. The Power of 5 Apples**

* The agent **loves** reaching exactly **5 apples**.
* From there, it knows which action can **guarantee a win**.
* So it tries to **steer the game** toward this number — even planning moves two steps ahead to get there.

Example:

* From 8 → take 3 → leave 5
* From 7 → take 2 → leave 5
* From 6 → take 1 → leave 5

It’s like the agent discovered a **safe zone** and does everything to land there.

#### **2. A Hidden Cycle Appears**

The moves reveal a kind of **cycle** or **rhythm** the agent has learned:

> "Force the opponent into a bad position that you've already mastered."

This behavior is **strategic**, not random. It shows the agent isn’t just reacting — it’s **building a staircase of victories**, one optimal step at a time.

#### **3. The Danger Zones**

Certain states (like **1**, **5**, or **9**) are almost always losses — and the agent knows it.

* From 9: no matter what move you pick, the opponent gets 6, 7, or 8 — all of which are **great positions**.
* From 1: any action ends the game, and **you lose**.

The agent **learned this the hard way** — by losing in these states repeatedly.

#### **4. A Mathematical Pattern: Modulo 4**

If you look closely, the agent seems to have discovered a **mathematical secret**:

> The best strategy is to leave the opponent with a number of apples such that
> **`apples_left % 4 == 1`** (like 9, 5 and 1).

This is the **winning position** in many "take-away" games. The agent figured this out **just from data** it itself has created by experiencing — no formulas, no instructions. It didn't even know what the target of the game is.

#### **5. Confidence from Repetition**

Some actions now have **perfect scores of +100 or -100**: This means the agent played those states **hundreds of times** and **always** won or **always** lost after certain actions.

That kind of confidence means the learning has **converged**, and the agent has a **strong belief** in what works best.

`````python
# === Training ===
ai = Player('AI')  # Acts as the AI agent
ai.train(10000)
`````

`````python
ai.plot_records()
`````

`````python
# Play with human
human = Player('Human')
board = Environment()

while not board.winner:
    board.show_board()
    p = ai if board.aiTurn else human
    move = p.decide(board)
    board.apply_action(move)
    print(f'{p.name} removed {move} apples')

print(board.winner, 'is the winner')
`````
